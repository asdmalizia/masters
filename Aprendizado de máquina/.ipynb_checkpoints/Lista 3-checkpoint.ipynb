# **Questão 1 — K-means unidimensional**

## **(a)**

Primeiro, ordenamos os pontos de dados em ordem crescente:

$$
x^{(1)} \le x^{(2)} \le \cdots \le x^{(N)}.
$$

Queremos provar que, em uma solução ótima do K-means unidimensional, cada cluster é formado por **pontos consecutivos** nessa ordenação.

---

### **Prova (por contradição)**

Suponha que exista uma atribuição ótima em que um cluster contenha pontos **não consecutivos**, ou seja,  
existem índices $i < j < l$ tais que $x^{(i)}$ e $x^{(l)}$ pertencem ao mesmo cluster $C_r$,  
mas $x^{(j)}$ pertence a outro cluster $C_s$.

Os centroides de cada cluster são as médias dos pontos atribuídos a eles:

$$
\mu_r = \frac{1}{|C_r|}\sum_{x \in C_r} x, 
\qquad
\mu_s = \frac{1}{|C_s|}\sum_{x \in C_s} x.
$$

Como $x^{(i)} < x^{(j)} < x^{(l)}$, temos $\mu_r < \mu_s$.  
Portanto, a distância de $x^{(j)}$ ao centro $\mu_s$ é **maior** que a distância ao centro $\mu_r$,  
violando a atribuição ótima (pois cada ponto deve ser atribuído ao centro mais próximo).

Assim, qualquer solução ótima deve ter a propriedade de que cada cluster contém **intervalos consecutivos** de pontos.  
Isto é, existem índices $1 = i_0 < i_1 < i_2 < \cdots < i_K = N$ tais que o cluster $j$ contém:

$$
C_j = \{x^{(i_{j-1}+1)}, x^{(i_{j-1}+2)}, \dots, x^{(i_j)}\}.
$$

---

## **(b)**

Desejamos agora desenvolver um algoritmo de **programação dinâmica** de complexidade $O(KN^2)$  
para o problema unidimensional.

---

### **Ideia geral**

Sabemos que cada cluster é formado por pontos consecutivos.  
Logo, basta determinar as **fronteiras** que separam os $K$ clusters, ou seja, os índices  
$i_1, i_2, \dots, i_{K-1}$.

Seja:

$$
\text{cost}(p,q) = \sum_{i=p}^{q} (x^{(i)} - \bar{x}_{p:q})^2,
$$

onde $\bar{x}_{p:q}$ é a média dos pontos entre $x^{(p)}$ e $x^{(q)}$.

Essa função representa o **erro quadrático total** (SSE) de agrupar os pontos $x^{(p)}, \dots, x^{(q)}$ em um cluster.

---

### **Recorrência dinâmica**

Definimos:

$$
DP[k][n] = \text{custo mínimo para particionar os primeiros } n \text{ pontos em } k \text{ clusters.}
$$

A relação de recorrência é:

$$
DP[k][n] = \min_{m < n} \{ DP[k-1][m] + \text{cost}(m+1, n) \}.
$$

- $DP[1][n] = \text{cost}(1, n)$ (todos os pontos em um único cluster),
- e o resultado final é $DP[K][N]$.

---

### **Cálculo eficiente de $\text{cost}(p,q)$**

Podemos calcular $\text{cost}(p,q)$ em tempo $O(1)$ com prefixos:

$$
S_1(i) = \sum_{t=1}^{i} x^{(t)}, \qquad S_2(i) = \sum_{t=1}^{i} (x^{(t)})^2.
$$

Assim:

$$
\bar{x}_{p:q} = \frac{S_1(q)-S_1(p-1)}{q-p+1},
$$

e

$$
\text{cost}(p,q) = S_2(q)-S_2(p-1)
 - \frac{(S_1(q)-S_1(p-1))^2}{q-p+1}.
$$

---

### **Algoritmo completo**

1. Ordene $x^{(1)}, \dots, x^{(N)}$.
2. Compute prefixos $S_1$ e $S_2$.
3. Inicialize $DP[1][n] = \text{cost}(1,n)$.
4. Para $k=2,\dots,K$:
   - Para $n=k,\dots,N$:
     - $DP[k][n] = \min_{m=k-1,\dots,n-1} \big[ DP[k-1][m] + \text{cost}(m+1,n) \big]$.
5. O resultado é $DP[K][N]$.

---

### **Complexidade**

O cálculo de $\text{cost}(p,q)$ é $O(1)$ após prefixos.  
A dupla iteração em $k$ e $n$ com varredura de $m$ gera custo total $O(KN^2)$.

---

### **Resumo**

- Cada cluster ótimo contém pontos consecutivos.
- A recorrência dinâmica permite encontrar a partição ótima em tempo polinomial.
- A versão 1D do K-means tem, portanto, **solução exata em $O(KN^2)$**, enquanto a versão multidimensional ($D>1$) é NP-difícil.

# **Questão 2 — Mistura Gaussiana e passo E do EM**

---

# **(a) O que o método EM está otimizando conceitualmente?**

No modelo de mistura Gaussiana com $K$ componentes, estamos assumindo que cada ponto $x_i$ foi gerado por uma das $K$ Gaussianas, mas **não sabemos qual**.

O método EM otimiza a **verossimilhança dos dados sob o modelo de mistura**:

$$
p(X \mid \pi, \mu, \sigma^2)
= 
\prod_{i=1}^n \sum_{k=1}^K 
\pi_k \, \mathcal{N}(x_i \mid \mu_k, \sigma_k^2).
$$

Como diretamente maximizar essa verossimilhança é difícil devido ao termo de soma dentro do produto, introduzimos variáveis latentes $z_i$ que indicam de qual Gaussiana o ponto veio.

O EM alterna entre:

---

### **E-step (Expectation step)**  
Computa as **responsabilidades**:

$$
\gamma_{ik} = p(z_i = k \mid x_i)
= \frac{\pi_k \, \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)}
       {\sum_{j=1}^K \pi_j \, \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)}.
$$

Elas representam a probabilidade de cada ponto pertencer a cada componente.

---

### **M-step (Maximization step)**  
Maximiza o valor esperado do log da verossimilhança completa, atualizando:

- pesos $\pi_k$  
- médias $\mu_k$  
- variâncias $\sigma_k^2$

---

### **Conceitualmente**, o EM está:

- estimando as variáveis ocultas (E-step),  
- usando essas estimativas para maximizar a verossimilhança (M-step).

Ele aumenta a verossimilhança a cada iteração até convergir.

---

# **(b) Passo E: cálculo das responsabilidades**

Temos os dados:

$$
X = \{1.0,\, 1.2,\, 0.8,\, 4.2,\, 3.9,\, 4.4\}.
$$

Parâmetros iniciais:

$$
\pi_1^{(0)} = \pi_2^{(0)} = 0.5,
\qquad
\mu_1^{(0)} = 1.0,\; \mu_2^{(0)} = 4.0,
$$

$$
(\sigma_1^2)^{(0)} = (\sigma_2^2)^{(0)} = 1.0.
$$

Cada responsabilidade é:

$$
\gamma_{i1}
= 
\frac{\pi_1 \, \mathcal{N}(x_i \mid 1, 1)}
     {\pi_1 \, \mathcal{N}(x_i \mid 1, 1)
     + \pi_2 \,\mathcal{N}(x_i \mid 4, 1)},
\qquad
\gamma_{i2} = 1 - \gamma_{i1}.
$$

Como $\pi_1 = \pi_2 = 1/2$, temos:

$$
\gamma_{i1} =
\frac{\mathcal{N}(x_i \mid 1, 1)}
     {\mathcal{N}(x_i \mid 1, 1) + \mathcal{N}(x_i \mid 4, 1)}.
$$

A densidade Gaussiana unidimensional é:

$$
\mathcal{N}(x \mid \mu, 1) 
= \frac{1}{\sqrt{2\pi}} 
\exp\!\left(-\frac{(x-\mu)^2}{2}\right).
$$

Como o denominador $\frac{1}{\sqrt{2\pi}}$ aparece em ambos os termos, podemos ignorá-lo:

$$
\gamma_{i1}
=
\frac{\exp\!\left(-\frac{(x_i-1)^2}{2}\right)}
{\exp\!\left(-\frac{(x_i-1)^2}{2}\right)
 + \exp\!\left(-\frac{(x_i-4)^2}{2}\right)}.
$$

A seguir calculamos para cada ponto.

---

## **Cálculo numérico das responsabilidades**

### Para $x_1 = 1.0$:

$$
\gamma_{11}
= 
\frac{e^{-(0)^2/2}}
     {e^{-(0)^2/2} + e^{-(3)^2/2}}
=
\frac{1}{1 + e^{-9/2}}
\approx 0.9889.
$$

---

### Para $x_2 = 1.2$:

     {e^{-(0.2)^2/2} + e^{-(2.8)^2/2}}
\approx
\frac{e^{-0.02}}{e^{-0.02} + e^{-3.92}}
\approx 0.9805.
$$

---

### Para $x_3 = 0.8$:

$$
\gamma_{31}
\approx
\frac{e^{-0.02}}{e^{-0.02} + e^{-5.12}}
\approx 0.9898.
$$

---

### Para $x_4 = 4.2$:

$$
\gamma_{41}
\approx
\frac{e^{-5.12}}{e^{-5.12} + e^{-0.02}}
\approx 0.0102.
$$

---

### Para $x_5 = 3.9$:

$$
\gamma_{51}
\approx
\frac{e^{-4.205}}{e^{-4.205} + e^{-0.005}}

$$

---

### Para $x_6 = 4.4$:

$$
\gamma_{61}
\approx

$$

## **Resumo — Responsabilidades**

|------|--------------:|--------------:|

Observamos que os três primeiros pontos têm probabilidade ≈ 1 de pertencer ao cluster da média 1.0, e os três últimos ao cluster da média 4.0.

# **(c) Passo M — Atualização dos parâmetros**

O passo M atualiza:

$$
N_k = \sum_{i=1}^n \gamma_{ik}, 
\qquad 
\pi_k = \frac{N_k}{n},
$$

$$
\mu_k = \frac{1}{N_k} \sum_{i=1}^n \gamma_{ik} x_i,
\qquad
\sigma_k^2 = \frac{1}{N_k} \sum_{i=1}^n \gamma_{ik} (x_i - \mu_k)^2.
$$

---

## **Cálculo de $N_1$ e $N_2$**

$$
N_1 = 0.9889 + 0.9805 + 0.9898 + 0.0102 + 0.0148 + 0.0067 \approx 2.991.
$$

$$
N_2 = 6 - N_1 = 3.009.
$$

Logo:

$$
\pi_1 = \frac{N_1}{6} \approx 0.4985,
\qquad
\pi_2 = \frac{N_2}{6} \approx 0.5015.
$$

Os pesos continuam praticamente iguais a 0.5.

---

## **Cálculo da nova média $\mu_1$**

$$
\mu_1 
= \frac{1}{N_1}
(0.9889\cdot 1.0 + 0.9805\cdot 1.2 + 0.9898\cdot 0.8 + 0.0102\cdot 4.2 + 0.0148\cdot 3.9 + 0.0067\cdot 4.4).
$$

Computando:

- $0.9889\cdot1.0 \approx 0.9889$
- $0.9805\cdot1.2 \approx 1.1766$
- $0.9898\cdot0.8 \approx 0.7918$

$$
\sum \gamma_{i1} x_i \approx 3.0873.

$$
\mu_1 \approx \frac{3.0873}{2.991} \approx 1.032.
$$

Os valores são pequenos porque os dados de cada cluster estão bem concentrados.

# **(d) Qual componente fica responsável por cada grupo?  
Por que o EM se comporta como clustering?**

Os três primeiros pontos $\{1.0, 1.2, 0.8\}$ têm responsabilidades muito próximas de 1 para o cluster 1.  
Os três últimos pontos $\{4.2, 3.9, 4.4\}$ têm responsabilidades muito próximas de 1 para o cluster 2.

Assim, **cada cluster “assume responsabilidade” por um grupo distinto dos dados**, exatamente como no K-means.

### Por que isso ocorre?

- A mistura Gaussiana maximiza a verossimilhança atribuindo alta probabilidade a pontos “perto” de cada média.
- Em regiões onde a Gaussiana 1 domina, $\gamma_{i1}\approx 1$ e $\gamma_{i2}\approx 0$.
- Em regiões onde a Gaussiana 2 domina, ocorre o oposto.

Mesmo sendo um método probabilístico, as responsabilidades convergem rapidamente para valores próximos de 0 ou 1 quando os clusters estão bem separados — levando a um comportamento idêntico ao de clustering determinístico.

---

# **(e) Por que o EM nunca reduz a verossimilhança?**

O EM **sempre aumenta (ou mantém) a verossimilhança** por construção, porque:

1. **E-step:**  
   Calcula o valor esperado do log da verossimilhança completa usando a distribuição posterior atual.  
   Isso sempre cria um *bound* inferior da verossimilhança.

2. **M-step:**  
   Maximiza esse bound em relação aos parâmetros.  
   Isso **não pode diminuir** o bound — apenas aumentá-lo ou mantê-lo.

3. Como a verossimilhança observada é sempre maior ou igual ao bound, o resultado é:

$$
\mathcal{L}(w^{(t+1)}) \ge \mathcal{L}(w^{(t)}).
$$

Portanto, o EM é um **algoritmo monotonicamente crescente em verossimilhança**.

# **Questão 3 — Relação entre MAP e Regressão Ridge**

Considere o modelo de regressão linear:

$$
y^{(i)} = w^\top x^{(i)} + \epsilon,
\qquad \epsilon \sim \mathcal{N}(0,\sigma^2),
$$

e, em forma matricial:

$$
y = Xw + \epsilon, 
\qquad y|X,w \sim \mathcal{N}(Xw,\sigma^2 I).
$$

Além disso, o vetor de parâmetros possui um prior Gaussiano:

$$
w \sim \mathcal{N}\!\left(0,\, \frac{2\sigma^2}{\lambda} I\right),
\qquad \lambda > 0.
$$

Queremos mostrar que o estimador MAP de $w$ é o mesmo que o estimador da **regressão ridge**, cuja solução é:

$$
w^\* = \arg\min_w \|y - Xw\|_2^2 + \lambda \|w\|_2^2.
$$

---

# **Solução**

## **1. Posterior proporcional ao produto da verossimilhança pelo prior**

A densidade posterior é:

$$
p(w|X,y) \propto p(y|X,w)\, p(w).
$$

### **Verossimilhança:**

$$
p(y|X,w)
\propto 
\exp\!\left(
-\frac{1}{2\sigma^2}\|y - Xw\|_2^2
\right).
$$

### **Prior Gaussiano:**

$$
p(w)
\propto 
\exp\!\left(
-\frac{1}{2} 
w^\top 
\left(\frac{\lambda}{2\sigma^2}I\right) 
w
\right)
=
\exp\!\left(
-\frac{\lambda}{4\sigma^2} \|w\|_2^2
\right).
$$

---

## **2. Log-posterior (negativo)**

Tomando o logaritmo e ignorando constantes independentes de $w$:

$$
-\log p(w|X,y) 
= 
\frac{1}{2\sigma^2}\|y - Xw\|_2^2 
+
\frac{\lambda}{4\sigma^2}\|w\|_2^2
\quad + \; \text{cte}.
$$

Multiplicamos tudo por $2\sigma^2$ (não afeta o minimizador):

$$
\arg\min_w 
\left[
\|y - Xw\|_2^2 
+ 
\frac{\lambda}{2} \|w\|_2^2
\right].
$$

Como $\lambda$ é arbitrário, podemos redefinir o hiperparâmetro da penalização como  
$\lambda_{\text{ridge}} = \lambda/2$, e obtemos:

$$
w_{\text{MAP}}
= 
\arg\min_w \left(
\|y - Xw\|_2^2 + 
\lambda_{\text{ridge}} \|w\|_2^2
\right).
$$

---

# **Conclusão**

A solução MAP coincide **exatamente** com a solução da regressão ridge, pois:

- o prior Gaussiano introduz uma penalização quadrática em $\|w\|_2^2$,
- o termo da verossimilhança gera o erro quadrático $\|y - Xw\|_2^2$,
- o produto dos dois leva à função objetivo da regressão ridge.

Portanto:

$$
\boxed{
w_{\text{MAP}} = w_{\text{ridge}}
}
$$

ou seja, **regressão ridge = estimativa MAP com prior Gaussiano isotrópico**.

# **Questão 4 — Aceleração de busca de vizinhos via SVD**

Considere um dataset com $n$ pontos em $\mathbb{R}^{1000}$, armazenados em uma matriz:

$$
A \in \mathbb{R}^{n \times 1000},
$$

cada linha sendo um ponto. Sabemos que $\operatorname{rank}(A) = 10$.  
Recebemos um *stream* de pontos $X = \{x^1, \dots, x^m\}$ e queremos, para cada $x^i$, encontrar os **5 pontos mais próximos** no dataset.

---

## **(a) Proximidade medida pelo produto interno**

A similaridade entre dois pontos é:

$$
\text{sim}(x, a_j) = x^\top a_j,
$$

e queremos os pontos com **maior produto interno**.

Como $A$ possui posto 10, aplicamos sua decomposição SVD truncada:

$$
A \approx U_{10} \Sigma_{10} V_{10}^\top,
$$

onde:

- $U_{10} \in \mathbb{R}^{n \times 10}$
- $\Sigma_{10} \in \mathbb{R}^{10 \times 10}$
- $V_{10} \in \mathbb{R}^{1000 \times 10}$.

Cada ponto da base pode ser representado no espaço reduzido de 10 dimensões:

$$
z_j = \Sigma_{10} U_{10}^{\top}(j,:), \qquad j=1,\dots,n.
$$

Da mesma forma, projetamos cada ponto do *stream*:

$$
x_{\text{proj}} = \Sigma_{10}^{-1} V_{10}^{\top} x.
$$

Como:

$$
x^\top a_j \approx x_{\text{proj}}^\top z_j,
$$

podemos computar **todos os produtos internos no espaço de 10 dimensões**, reduzindo o custo de $O(1000n)$ para $O(10n)$.  

Depois, basta selecionar os 5 maiores valores.

---

## **(b) Proximidade medida pela distância Euclidiana**

Agora queremos minimizar:

$$
\|x - a_j\|_2.
$$

Usamos novamente a SVD truncada:

$$
A \approx U_{10} \Sigma_{10} V_{10}^\top.
$$

Cada ponto do dataset é projetado como:

$$
z_j = \Sigma_{10} U_{10}^{\top}(j,:), 
$$

e cada ponto novo:

$$
x_{\text{proj}} = \Sigma_{10}^{-1} V_{10}^{\top} x.
$$

Como a SVD fornece a **melhor aproximação rank-10** em norma de Frobenius, vale:

$$
\|x - a_j\|_2^2 \approx \|x_{\text{proj}} - z_j\|_2^2 + \text{constante},
$$

onde a constante não depende de $j$ e não afeta a ordenação dos vizinhos mais próximos.

Assim, podemos realizar a busca de vizinhos no espaço reduzido de dimensão 10:

$$
j^* = \arg\min_j \|x_{\text{proj}} - z_j\|_2.
$$

Isso reduz drasticamente o custo computacional de cada consulta:

- distância no espaço original: $O(1000n)$  
- distância no espaço reduzido: $O(10n)$

e a precisão permanece alta graças à baixa-rankidade da matriz de dados.

---

## **Resumo**

### Produto interno (similaridade)
- Projetar todos os pontos para o subespaço de dimensão 10.
- Calcular produtos internos no espaço reduzido.
- Buscar os 5 maiores.

### Distância Euclidiana
- Projetar pontos para o mesmo subespaço.
- Calcular distâncias no espaço reduzido.
- Buscar os 5 menores.

### Ganho
$$
\text{Complexidade cai de } O(1000n) \text{ para } O(10n) \text{ por consulta.}
$$

A SVD aproveita o fato de que os dados têm **posto 10**, realizando buscas eficientes em um espaço de dimensão muito menor.
