# Estimação do modelo

```python
def negloglik(params, data):

    kappa, sigma, xi = params
    if kappa <= 0 or sigma <= 0: # restrições
        return np.inf

    model = NaveauModelI(kappa, sigma, xi)
    logf = model.logpdf(data)

    if np.any(~np.isfinite(logf)):
        return np.inf

    return -np.sum(logf)
```

```python
# remoção de observações nulas
series_treated = series[series>0]
series_molhado_treated = series_molhado[series_molhado>0]
series_seco_treated = series_seco[series_seco>0]
```

```python
# estimação all data
init = [1.0, 1.0, 0.1]

res = minimize(
    negloglik,
    x0=init,
    args=(series_treated,),
    method="L-BFGS-B",
    bounds=[
        (1e-6, None),   # kappa
        (1e-6, None),   # sigma
        (-1, 5)         # xi
    ]
)
print("Parâmetros estimados do naveau m1 (all data): \nkappa={:.2f}, \nsigma={:.2f}, \nxi={:.2f}".format(*res.x))
kappa_all, sigma_all, xi_all = res.x

# estimação com estação molhada
init = [1.0, 1.0, 0.1]

res = minimize(
    negloglik,
    x0=init,
    args=(series_molhado_treated,),
    method="L-BFGS-B",
    bounds=[
        (1e-6, None),   # kappa
        (1e-6, None),   # sigma
        (-1, 5)         # xi
    ]
)
print("\nParâmetros estimados do naveau m1 (molhado): \nkappa={:.2f}, \nsigma={:.2f}, \nxi={:.2f}".format(*res.x))
kappa_molhado, sigma_molhado, xi_molhado = res.x

# estimação com estação seca
init = [1.0, 1.0, 0.1]

res = minimize(
    negloglik,
    x0=init,
    args=(series_seco_treated,),
    method="L-BFGS-B",
    bounds=[
        (1e-6, None),   # kappa
        (1e-6, None),   # sigma
        (-1, 5)         # xi
    ]
)
print("\nParâmetros estimados do naveau m1 (seco): \nkappa={:.2f}, \nsigma={:.2f}, \nxi={:.2f}".format(*res.x))
kappa_seco, sigma_seco, xi_seco = res.x
```
