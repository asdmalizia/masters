```python
import numpy as np
import matplotlib.pyplot as plt

# parâmetros
mu = 0      # localização
tau = 1     # escala

# domínio
x = np.linspace(-10, 10, 500)

# função
p = 1/2 + (1/np.pi) * np.arctan((x - mu) / tau)

# plot
plt.figure(figsize=(6,4))
plt.plot(x, p, label=r'$p_{\mu,\tau}(x)$')

plt.title('CDF da distribuição de Cauchy')
plt.xlabel('x')
plt.ylabel('p(x)')
plt.grid(True)
plt.legend()

plt.show()
```
