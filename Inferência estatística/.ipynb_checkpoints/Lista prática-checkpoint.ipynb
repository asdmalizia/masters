## item d)
Construa um intervalo de confiança para o parâmetro, aplique-o para cada sub-
amostra e contabilize o porcentual de vezes que o IC contém o parâmetro de
interesse.

---
$\textbf{Resolução}$   

Inicialmente, foi desenvolvido o cálculo do intervalo de confiança para o estimador de $\theta$ usando uma estatística Z.

![Gráfico bonito](img/3d.jpg)

![Gráfico bonito](img/3d_ii.jpg)

Usando os valores críticos da estatística Z com nível de confiança $\alpha=5%$, os limites inferior e superior do intervalo desenvolvido foi calculado para cada sub-amostra. Os valores obtidos foram salvos em vetores de dimensão 100. 

Em seguida, o valor populacional do parâmetro foi comparado com os limites do intervalo de cada sub-amostra, armazenando os resultados em um vetor booleano de dimensão 100: 1 representa sub-amostras em que o parâmetro ultrapassou os limites do intervalo de confiança calculado.

```python
def calcula_ic(α, samples):
    z_alpha = stats.norm.ppf(1-α/2)
    lower_bound = np.mean(samples, axis=1) / (
        2 + z_alpha/(
            pow(samples.shape[1]/2, 0.5)
        )
    ) 
    upper_bound = np.mean(samples, axis=1) / (
        2 - z_alpha/(
            pow(samples.shape[1]/2, 0.5)
        )
    ) 
    return lower_bound, upper_bound

# cálculo do IC
α = 0.05
lower_bound, upper_bound = calcula_ic(α, samples)
out_of_bounds = (θ < lower_bound) | (θ > upper_bound)
```

Os resultados obtidos foram ilustrados em um gráfico com o intervalo de confiança calculado para cada sub-amostra, destacando aqueles em que o parâmetro não se encaixou. 

Pelo gráfico, é possível observar que em 5 das 100 amostras o parâmetro $\theta$ não estava contido no intervalo de confiança, o que corresponde ao nível de significância de 5% escolhido pro cálculo do intervalo.

## item e)
Desenvolva um teste de hipóteses para determinar se a proporção observada
no item (e) é igual à prevista (o nível de confiança).

---
$\textbf{Resolução}$   

Para desenvolver o teste, foi considerada uma variável aleatória X, que assume o valor 1 quando o parâmetro está fora do intervalo de confiança calculado e zer, caso contrário. Essa variável possui distribuição de Bernoulli com probabilidade p igual ao nível de confiança escolhido (5%).

Dessa forma, o teste para identificar a proporção de estimativas fora do intervalo de confiança equivale a um teste de proporção de uma amostra aleatória de X, onde a hipótese nula é a de que a proporção é igual ao nível de confiança. Esse teste é feito com uma estatística Z.

![Gráfico bonito](img/3e.jpg)

Usando a estatística de teste desenvolvida para a proporção, foi calculado o seu valor, o valor crítico da estatística Z para o nível de confiança 95% e o p-valor. 

Segundo os resultados obtidos, a estatística de teste não está na região de rejeição do teste. Além disso, o p-valor é alto. Portanto, o teste falha em rejeitar a hipótese nula de que a proporção é igual a 5%.

```python
def calcula_estatistica_de_teste(sample, p0, α=0.05):
    estimativa = np.mean(sample)
    z = (estimativa - p0) / pow(
            p0*(1-p0)/len(sample), 0.5
        )
    z_alpha = stats.norm.ppf(1-α/2)
    p_value = 2*(1 - stats.norm.cdf(x=abs(z)))
    return z, z_alpha, p_value

z, z_alpha, p_value = calcula_estatistica_de_teste(out_of_bounds, p0=α, α=0.05)
print('Estatística de teste: {:.2f} \nValor crítico: {:.2f} \np-valor: {:.2f}'.format(z, z_alpha, p_value))
```

## item f)
Desenvolva um teste para verificar se as variâncias são iguais.

```python
def calcula_estatistica_de_teste(samples, α=0.05):
    μ1_hat, μ2_hat = np.mean(samples[0]), np.mean(samples[1])
    n1, n2 = samples[0].shape[0], samples[1].shape[0]
    
    S1 = np.sum(
        np.power(samples[0] - μ1_hat, 2)
    ) / (n1-1)
    S2 = np.sum(
        np.power(samples[1] - μ2_hat, 2)
    ) / (n2-1)
    
    F = S1/S2
    F_alpha = (
        stats.f.ppf(q = α/2, dfn = n1-1, dfd = n2-1),
        stats.f.ppf(q = 1-α/2, dfn = n1-1, dfd = n2-1)
    )
    return F, F_alpha
```

```python
F, bounds = calcula_estatistica_de_teste(samples, α=0.05)
F, bounds
```

## item h) 
Plote a curva característica do teste do item (a) como uma função de $ \lambda = \frac{\sigma^2_1}{\sigma^2_2}$

```python
def calcula_erro_tipo_ii(n1, n2, λ, α = 0.05):
    lower = stats.f.ppf(q = α/2, dfn = n1-1, dfd = n2-1)
    upper = stats.f.ppf(q = 1-α/2, dfn = n1-1, dfd = n2-1)
    
    lower_star, upper_star = lower/λ, upper/λ
    β = (
        stats.f.cdf(x=upper_star, dfn=n1-1, dfd=n2-1) -
        stats.f.cdf(x=lower_star, dfn=n1-1, dfd=n2-1)
    )
    return β
```

```python
calcula_erro_tipo_ii(n1=samples[0].shape[0], n2=samples[1].shape[0], λ=1)
```

## item l) 
Escreva um teste de razão de verossimilhança para igualdade das médias.   
i. Calcule uma expressão para a estatística de teste   
ii. Determine a regra de rejeição considerando a distribuição assintótica da
estatística de teste.

```python
def calcula_estatística_de_teste(samples, μ0, α=0.05):
    
    n1, n2 = len(samples[0]), len(samples[1])
    μ1, μ2 = np.mean(samples[0]), np.mean(samples[1])
    λ1 = pow(
        np.sum((samples[0] - μ1)**2)/
         np.sum((samples[0] - μ0)**2), 
        (n1/2)
    )
    λ2 = pow(
        np.sum((samples[1] - μ2)**2)/
         np.sum((samples[1] - μ0)**2), 
        (n2/2)
    )
    λ = λ1*λ2

    χ = -2*np.log(λ)
    χ_alpha = (
        stats.chi2.ppf(q = α/2, df=2),
        stats.chi2.ppf(q = 1-α/2, df=2)
    )
    if χ < χ_alpha[0] or χ > χ_alpha[1]: reject = True
    else: reject = False
        
    return χ, χ_alpha, reject
```

```python
μ1 = np.mean(samples[0])
calcula_estatística_de_teste(samples, μ0=μ1, α=0.05)
```

## b) 
Apresente os estimadores não viesados de $ \beta_0, \beta_1 $ e $\sigma^2$ . Explique como são obtidos. Use um mínimo de equações.

```python
def calcula_estimadores(Y, X):
    β_hat = np.linalg.inv(X.T @ X) @ (X.T @ Y)

    n, p = len(Y), len(β_hat)
    ε2 = (Y - X @ β_hat).T @ (Y - X @ β_hat)
    σ2_hat = ε2/(n - p)
    
    return β_hat[0], β_hat[1], σ2_hat**0.5
```

## e) 
Aplique os estimadores do item b e calcule o R2. Anote o resultado.

```python
β0_hat, β1_hat, σ_hat = calcula_estimadores(Y, X)
β0_hat, β1_hat, σ_hat
```

```python
def calcula_R2(β0, β1, Y, x):
    y_mean = np.mean(Y)
    Y_hat = β0 + β1*x
    
    SSr = np.sum((Y_hat - y_mean)**2)
    Syy = np.sum((Y - y_mean)**2)
    return SSr/Syy, Y_hat
```

```python
R2, _ = calcula_R2(β0_hat, β1_hat, Y, xn)
R2
```

## f) 
Repita as etapas 1.d e 1.e K vezes, K > 100.

```python
K = 200
Y_hat = np.zeros((K,n))
R2 = np.zeros(K)

for i in range(K):
    ε = rng.normal(loc=0, scale=σ, size=n)
    Y = β0 + β1*xn + ε
    β0_hat, β1_hat, σ_hat = calcula_estimadores(Y, X)
    R2[i], Y_hat[i,:] = calcula_R2(β0_hat, β1_hat, Y, xn)
```

## g)

```python
fig, ax = plt.subplots(figsize=(8,3))

for i in range(K):
    ax.scatter(xn, Y_hat[i,:], alpha=0.4, s=8)
    if i<20:
        ax.plot(xn, Y_hat[i,:], linestyle=':', linewidth=0.5)

ax.set_title('Simulações')
plt.show()
```
