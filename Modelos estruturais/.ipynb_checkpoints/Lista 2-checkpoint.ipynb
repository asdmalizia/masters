# Funções de estimação

```python
def KalmanFilter(Y, T, Z, Q, H, C, D, x0, P0):
    
    n, p, m = len(Y), Y.shape[1], T.shape[0]
    Z_t, H_t = Z.copy(), H.copy()
    
    # inicialização dos arrays
    x_hat = np.ones((n, m, 1))
    x_hat_minus = np.ones((n, m, 1))
    P_minus_t = np.zeros((n, m, m))
    P = np.zeros((n, m, m))
    F_t = np.zeros((n, p, p))
    v_t = np.zeros((n, p, 1))
    
    # valores inicias
    x_hat[0] = x0
    x_hat_minus[0] = x0
    P[0] = P0
    P_minus_t[0] = P0
    F_t[0] = H_t[0]
    log_likelihood = -n/2 * np.log(2 * np.pi)

    # Kalman filter
    for iteration in range(1, n):
        # previsão
        x_hat_minus[iteration] = T @ x_hat[iteration - 1] + C        
        P_minus = np.array(T @ P[iteration - 1] @ T.T + Q)
        P_minus = P_minus.reshape((m, m))
        P_minus_t[iteration] = P_minus

        # muda Zt, Ht e inovação para observações faltantes    
        if np.isnan(Y[iteration]).any():
            Z_t[iteration] = np.zeros(Z_t[iteration].shape)
            H_t[iteration] = np.eye(p)
            eta_t = np.zeros((p,1))
        else:
            eta_t = Y[iteration].reshape(p, 1) - Z_t[iteration] @ x_hat_minus[iteration] - D
        
        # inovação e variança da inovação  
        v_t[iteration] = eta_t
        F_t[iteration] = Z_t[iteration] @ P_minus_t[iteration] @ Z_t[iteration].T + H_t[iteration]
        covariance_determinant = np.linalg.det(F_t[iteration])
        if abs(covariance_determinant) == 0 or str(covariance_determinant).lower() == "nan":
            return -1e10, x_hat_minus, x_hat, P, v_t
        
        # atualiza log-likelihood
        log_like = np.log(np.linalg.det(F_t[iteration])) + eta_t.T @ np.linalg.inv(F_t[iteration]) @ eta_t
        log_likelihood -= 0.5 * log_like[0][0]

        # atualização
        Kalman_gain = P_minus_t[iteration] @ Z_t[iteration].T @ np.linalg.inv(F_t[iteration])
        x_hat[iteration] = x_hat_minus[iteration] + Kalman_gain @ eta_t
        P[iteration] = P_minus_t[iteration] - Kalman_gain @ Z_t[iteration] @ P_minus_t[iteration]

    return log_likelihood, x_hat_minus, x_hat, P, v_t
```

```python
def KalmanSmoother(Y, T, Z, Q, H, C, D, x0, P0):
    # roda o filtro forward
    log_likelihood, x_hat_minus, x_hat, P, v_t = KalmanFilter(Y, T, Z, Q, H, C, D, x0, P0)

    n = len(Y)
    m = T.shape[0]
    p = Y.shape[1]

    r = np.zeros((n+1, m, 1))     # vetores r_k
    N = np.zeros((n+1, m, m))     # matrizes N_k
    x_hat_smooth = np.zeros((n, m, 1))
    P_smooth = np.zeros((n, m, m))

    # backward
    for iteration in range(n-1, -1, -1):
        
        # muda Zt e Ht para observações faltantes    
        if np.isnan(Y[iteration]).any():
            Z[iteration] = np.zeros(Z[iteration].shape)
            H[iteration] = np.eye(p)
        
        # variâncias da inovação
        F_t = Z[iteration] @ P[iteration] @ Z[iteration].T + H[iteration]
        F_t_inv = np.linalg.inv(F_t)

        # ganho
        K_t = P[iteration] @ Z[iteration].T @ F_t_inv
        L_t = T - K_t @ Z[iteration]

        # acumula r e N
        if np.isnan(Y[iteration]).any():
            r[iteration] = L_t.T @ r[iteration+1]
        else:
            r[iteration] = Z[iteration].T @ F_t_inv @ v_t[iteration] + L_t.T @ r[iteration+1]
        N[iteration] = Z[iteration].T @ F_t_inv @ Z[iteration] + L_t.T @ N[iteration+1] @ L_t

        # estimativas suavizadas
        x_hat_smooth[iteration] = x_hat[iteration] + P[iteration] @ r[iteration]
        P_smooth[iteration] = P[iteration] - P[iteration] @ N[iteration] @ P[iteration]

    return log_likelihood, x_hat_smooth, x_hat_minus, x_hat, P, P_smooth, v_t
```

```python
def compute_aic(log_likelihood, n_obs, P0, num_variances):
    q = np.sum(np.diag(P0) >= 1e6) # estados iniciais difusos
    aic = (1 / n_obs) * (-2 * n_obs * log_likelihood + 2 * (q + num_variances))
    return aic
```

# 1) Sazonalidade com trigonométricos
$$
\begin{align*}
y_t &= \mu_t + \sum_{j=1}^{n_h} \gamma_{j,t} + \varepsilon_t, \\[0.3em]
\mu_{t+1} &= \mu_t + \beta_t + \eta_{t}, \\[0.3em]
\beta_{t+1} &= \beta_t + \zeta_{t}, \\[0.3em]
\begin{bmatrix}\gamma_{j,t+1}\\ \gamma_{j,t+1}^*\end{bmatrix} &=
\begin{bmatrix}\cos\lambda_j & \sin\lambda_j \\ -\sin\lambda_j & \cos\lambda_j\end{bmatrix}
\begin{bmatrix}\gamma_{j,t}\\ \gamma_{j,t}^*\end{bmatrix}
+
\begin{bmatrix}\omega_{j,t}\\ \omega_{j,t}^*\end{bmatrix}, \quad j=1,\ldots,n_h-1, \\[0.5em]
\end{align*}
$$

```python
def neg_log_likelihood(theta, Y, Z, T, C, D, x0, P0):
    sigma_eps, sigma_eta, sigma_zeta, sigma_omega = theta

    m = T.shape[0]
    H = np.array([[sigma_eps**2]])
    H_t = np.tile(H, (n, 1, 1))
    
    Q = np.zeros((m, m))
    Q[0, 0] = sigma_eta**2 # nivel
    Q[1, 1] = sigma_zeta**2 # inclinação
    Q[2:, 2:] = (sigma_omega**2) * np.eye(m - 2) # sazonalidade

    log_like, *_ = KalmanFilter(Y, T, Z, Q, H_t, C, D, x0, P0)
    return -log_like
```

```python
# n_harm = 6
model_params = []
for n_harm in range(1,7):
    
    n = series.shape[0]
    m = 2 + 2*n_harm - 1 # mu, beta, gamma 1 a n, gamma* 1 a n-1
    
    Z = np.zeros((1, m))
    Z[0, 0] = 1  # mu
    Z[0, 2::2] = 1  # pega todas as gammas e ignora os gammas*
    Z_t = np.tile(Z, (n, 1, 1)) # expande Z no tempo com Z fixo
    
    T = np.zeros((m, m))
    T[0, 0] = 1   # mu_t = mu_{t-1} + beta_{t-1}
    T[0, 1] = 1
    T[1, 1] = 1   # beta_t = beta_{t-1}
    
    # blocos sazonais C_j (2x2)
    for j in range(n_harm - 1):
        angle = np.pi * (j + 1) / 6
        Cj = np.array([
            [np.cos(angle), np.sin(angle)],
            [-np.sin(angle), np.cos(angle)]
        ])
        row = 2 + 2 * j
        T[row:row+2, row:row+2] = Cj
    T[-1, -1] = -1 # Último harmônico
    
    C = np.zeros((m, 1))
    D = np.zeros((1, 1))
    
    # inicialização com alta variancia para componentes não etacionários
    x0 = np.zeros((m, 1))
    P0 = np.eye(m)
    P0[0, 0] = 1e6  # mu_t
    P0[1, 1] = 1e6  # beta_t

    model_params.append((Z_t, T, C, D, x0, P0))

Y = series.values
```

## d) Estimação do número de harmônicos
Após estimar o modelo com 1, 2, 3, 4, 5 e 6 hamônicos trigonométricos, o modelo que apresentou menor AIC foi o de 4 harmônicos sazonais.

```python
# otimização dos parâmetros pela máxima verossimilhança
bounds_= [(1e-6, None)] * 4
initial_guess = np.array([10, 0.05, 0.05, 0.05])

est_models = []
for i in range(6):
    Z_t, T, C, D, x0, P0 = model_params[i]
    result = minimize(
        neg_log_likelihood,
        x0 = initial_guess,
        args = (Y, Z_t, T, C, D, x0, P0),
        method = "L-BFGS-B",
        bounds = bounds_,
        tol = 1e-6,
        options = {'maxiter': 200},
    )
    est_models.append({
        'params': result.x,
        'log_likelihood': -result.fun,
        'aic': compute_aic(-result.fun, n, P0, 4)
    })
    print(f'\n--------------------------- Modelo com {i+1} harmônicos ---------------------------')
    print('Parâmetros:', est_models[i]['params'])
    print('Máxima verossimilhança:', est_models[i]['log_likelihood'])
    print('AIC:', est_models[i]['aic'])
```

## e) Estimativas das variâncias
Para o modelo com menor AIC e 4 harmônicos, as variâncias estimadas da componente da inclinação e do slope foram iguais a zero. Esse resultado indica que essas coponentes são determinísicas.

Já a sazonalidade, é uma componente estocástica e possui o desvio igual a 3,6.

```python
# estimação do modelo ótimo
Z_t, T, C, D, x0, P0 = model_params[3] # AIC mínimo
sigma_eps, sigma_eta, sigma_zeta, sigma_omega = est_models[3]['params'] # parâmetros estimados

m = T.shape[0]
H = np.array([[sigma_eps**2]])
H_t = np.tile(H, (n, 1, 1))

Q = np.zeros((m, m))
Q[0, 0] = sigma_eta**2
Q[1, 1] = sigma_zeta**2
Q[2:, 2:] = sigma_omega**2 * np.eye(m - 2)
```

# 2) Sazonalidade com dummies

$$
\begin{align*}
y_t &= \mu_t + \gamma_t + \varepsilon_t \\
\mu_t &= \mu_{t-1} + \beta_{t-1} + \eta_t \\
\beta_t &= \beta_{t-1} + \zeta_t \\
\gamma_t &= D_t^\top S_t \\
S_t &= S_{t-1} + \omega_t
\end{align*}
$$


```python
def neg_log_likelihood(theta, Y, Z, T, C, D, x0, P0):
    sigma_eps, sigma_eta, sigma_zeta, sigma_omega = theta

    m = T.shape[0]
    H = np.array([[sigma_eps**2]])
    H_t = np.tile(H, (n, 1, 1))
    
    Q = np.zeros((m, m))
    Q[0, 0] = sigma_eta**2 # nivel
    Q[1, 1] = sigma_zeta**2 # inclinação
    I, i = np.eye(12), np.ones((12, 1))
    Sigma_star = sigma_omega * (I - (1/12) * i @ i.T)
    Q[2:, 2:] = Sigma_star # sazonalidade

    log_like, *_ = KalmanFilter(Y, T, Z, Q, H_t, C, D, x0, P0)
    return -log_like
```

```python
n = series.shape[0]
m = 14 # mu, beta, D1 a D12

Z_t = np.zeros((n, 1, m))
for t in range(n):
    month = series.index[t].month
    Z_t[t, 0, 0] = 1  # mu_t
    Z_t[t, 0, 2 + (month - 1)] = 1  # Dummies sazonais

T = np.eye(14)
T[0, 1] = 1  # mu_t = mu_{t-1} + beta_{t-1}

C = np.zeros((m, 1))
D = np.zeros((1, 1))

# inicialização com alta variancia para componentes não etacionários
x0 = np.zeros((m, 1))
P0 = np.eye(m) * 10
P0[0, 0] = 1e7  # mu_t
P0[1, 1] = 1e7  # beta_t

Y = series.values
```

```python
bounds_= [(1e-6, None)] * 4
initial_guess = np.array([10, 0.05, 0.05, 0.05])
parameters = [initial_guess.copy()]

callback_function.iteration = 1

result = minimize(
    neg_log_likelihood,
    x0 = initial_guess,
    args = (Y, Z_t, T, C, D, x0, P0),
    callback = callback_function,
    method = "L-BFGS-B",
    bounds = bounds_,
    tol = 1e-6,
    options = {'maxiter': 200},
)
result
```

## j) Capacidade preditiva

```python
def rolling_k_kalman_forecasts_all_steps(x_hat_smooth, P_smooth, T, Q, C, D, Z_t, H_t, max_k):
    n = len(x_hat_smooth)
    p = H_t.shape[1]
    forecasts = np.full((n, max_k, p), np.nan)

    for t in range(n):
        x_pred = x_hat_smooth[t].copy()
        P_pred = P_smooth[t].copy()
        max_h_for_t = min(max_k, n - t - 1)  # não ultrapassar o fim da série

        for h in range(max_h_for_t):
            x_pred = T @ x_pred + C
            P_pred = T @ P_pred @ T.T + Q
            y_pred = (Z_t[t + h + 1] @ x_pred + D).reshape(p)
            forecasts[t, h, :] = y_pred

    return forecasts
```

```python
def rolling_forecast_kalman(series, x_hat_smooth, P_smooth, T, Q, C, D, Z_t, H_t, max_k=24):
    n = len(series)
    idx = series.index
    forecasts = rolling_k_kalman_forecasts_all_steps(
        x_hat_smooth, P_smooth, T, Q, C, D, Z_t, H_t, max_k
    )

    cols = pd.MultiIndex.from_product([['Kalman'], [f'{h+1} passos' for h in range(max_k)]])
    df_kalman = pd.DataFrame(index=idx, columns=cols)

    for t in range(n):
        df_kalman.loc[idx[t]] = forecasts[t, :, 0]

    return df_kalman
```

```python
def rolling_forecast_ets(series, max_k=24, min_train=246):
    cols = pd.MultiIndex.from_product([['ETS'], [f'{h+1} passos' for h in range(max_k)]])
    df_ets = pd.DataFrame(index=series.index, columns=cols)

    for i in range(min_train, len(series)):
        train = series.iloc[:i]
        try:
            model = ETSModel(train, error='add', trend='add', seasonal='add',
                             damped_trend=True, seasonal_periods=12)
            fit = model.fit(disp=False)
            preds = fit.forecast(max_k)
            df_ets.loc[series.index[i]] = preds.values
        except Exception as e:
            pass

    return df_ets
```

```python
def rolling_forecast_arima(series, max_k=24, min_train=246):
    cols = pd.MultiIndex.from_product([['ARIMA'], [f'{h+1} passos' for h in range(max_k)]])
    df_arima = pd.DataFrame(index=series.index, columns=cols)

    for i in range(min_train, len(series)):
        train = series.iloc[:i]
        try:
            model = auto_arima(train, seasonal=True, m=12,
                               information_criteria='aicc', error_action='ignore', trace=False)
            preds = model.predict(n_periods=max_k)
            df_arima.loc[series.index[i]] = preds.values
        except Exception as e:
            pass

    return df_arima
```

```python
start = datetime.now()
df_ets = rolling_forecast_ets(series['Brasilia'])
df_arima = rolling_forecast_arima(series['Brasilia'])
df_kalman = rolling_forecast_kalman(series['Brasilia'], x_hat_smooth, P_smooth, T, Q, C, D, Z_t, H_t)
df_forecasts = pd.concat([df_kalman, df_ets, df_arima], axis=1)

df_forecasts = df_forecasts.astype(float)
end = datetime.now()
print(end-start)
```
