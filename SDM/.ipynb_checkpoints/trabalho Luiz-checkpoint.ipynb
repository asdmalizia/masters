# Modelos GAS Hurdle

## 1. GAS Hurdle Poisson (GHP)

Diferentemente das distribuições Poisson e Binomial Negativa (NB) padrão — nas quais a probabilidade de observar zero é função da média e variância —, nas distribuições aumentadas para zeros (zero‑augmented), como a hurdle Poisson, a probabilidade de zero surge de um processo separado.

Na hurdle Poisson (HP), primeiro ocorre um **ensaio de Bernoulli**, que gera um zero com probabilidade  
$$
\pi_{t|t-1}.
$$

Se ocorrer “falha” no Bernoulli, o valor positivo é amostrado de uma **Poisson truncada em zero**.

A função de massa de probabilidade da HP dinâmica é:

$$
p(y_t \mid \lambda_{t|t-1}, \pi_{t|t-1}) =
\begin{cases}
\pi_{t|t-1}, & y_t = 0, \\[8pt]
(1 - \pi_{t|t-1})
\dfrac{\lambda_{t|t-1}^{y_t} e^{-\lambda_{t|t-1}}}{y_t! \left(1 - e^{-\lambda_{t|t-1}}\right)}, 
& y_t > 0.
\end{cases}
\tag{15}
$$

com  
- $\lambda_{t|t-1} > 0$,  
- $\pi_{t|t-1} \in (0,1)$.

### 1.1 Média e Variância

**Média:**

$$
\mathbb{E}_{t-1}[y_t] = 
\frac{\lambda_{t|t-1}}{1 - e^{-\lambda_{t|t-1}}}(1 - \pi_{t|t-1})
$$

**Variância:**

$$
\operatorname{Var}_{t-1}[y_t] =
\lambda_{t|t-1}(1 - \pi_{t|t-1})
\frac{1}{1 - e^{-\lambda_{t|t-1}}}
\left[
1 +
\frac{\lambda_{t|t-1} - \pi_{t|t-1}\lambda_{t|t-1}}{1 - e^{-\lambda_{t|t-1}}}
-
\frac{\lambda_{t|t-1}^2}{\exp(\lambda_{t|t-1})(1 - e^{-\lambda_{t|t-1}})^2}
\right]
\tag{16}
$$

### 1.2 Dinâmica da Probabilidade de Zero

A probabilidade $\pi_{t|t-1}$ é dinâmica e evolui via mecanismo GAS:

$$
h_\pi(\pi_{t|t-1})
=
\delta
+
\beta\, h_\pi(\pi_{t-1|t-2})
+
\rho_2\;\tilde{s}_{\pi,t-1|t-2},
\qquad |\beta| < 1,\; \rho_2 > 0
\tag{17}
$$

com $h_\pi(\cdot)$ um link adequado (usualmente logit).

### 1.3 Score Reparametrizado Escalado

Para a parte Bernoulli:

$$
\tilde{s}_{y_t = 0} =
\begin{bmatrix}
1 - \pi_{t|t-1} \\
0
\end{bmatrix}
\tag{18}
$$

Para a parte Poisson truncada:

$$
\tilde{s}_{y_t > 0} =
\begin{bmatrix}
-\pi_{t|t-1} \\
y_t - 
\lambda_{t|t-1}
\frac{\lambda_{t|t-1} e^{-\lambda_{t|t-1}}}{1 - e^{-\lambda_{t|t-1}}}
\end{bmatrix}
\tag{19}
$$

Essa formulação permite previsão k‑passos‑à‑frente totalmente probabilística sem necessidade de soluções *ad hoc*.

### 1.4 Simulação

Abaixo simularemos o GHP ............

```python
from pathlib import Path

import numpy as np
import pandas as pd
import datetime as dt

from scipy.optimize import minimize
from scipy.special import expit, logit, gammaln
from scipy.stats import probplot, norm, poisson, nbinom, uniform

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def simulate_ghp(
    n_obs, 
    params, 
    seed=None,
    lambda_max=50, 
    loglam_bounds=(-5, 4), 
    logitpi_bounds=(-6, 6)
):
    """
    Simula uma série temporal a partir de um modelo GAS Hurdle Poisson (GHP) com
    reparametrização completa: dinâmica para log(λ_t) e para logit(π_t), com scores reparametrizados.

    Parâmetros:
    - n_obs: número de observações
    - params: dicionário com parâmetros fixos:
        {'omega1':..., 'alpha1':..., 'beta1':...,
         'omega2':..., 'alpha2':..., 'beta2':...}
    - seed: semente para reprodutibilidade

    Retorna:
    - y: série simulada
    - lambda_t: valores de λ_t ao longo do tempo
    - pi_t: valores de π_t ao longo do tempo
    """

    if seed is not None:
        np.random.seed(seed)

    # Dinâmica para log(λ_t)
    omega1 = params['omega1']
    alpha1 = params['alpha1']
    beta1  = params['beta1']

    # Dinâmica para logit(π_t)
    omega2 = params['omega2']
    alpha2 = params['alpha2']
    beta2  = params['beta2']

    # Estados latentes iniciais
    f1 = omega1  # log(λ)
    f2 = omega2  # logit(π)

    y = np.zeros(n_obs, dtype=int)
    lambda_t = np.zeros(n_obs)
    pi_t = np.zeros(n_obs)

    for t in range(n_obs):
        lam = np.exp(f1)
        logit_pi = f2
        pi = expit(logit_pi)

        # Clipping por segurança
        lam = np.clip(lam, 1e-6, lambda_max)
        logit_pi = np.clip(logit_pi, *logitpi_bounds)
        pi = np.clip(pi, 1e-6, 1 - 1e-6)

        lambda_t[t] = lam
        pi_t[t] = pi

        u = np.random.rand()
        if u < pi:
            y[t] = 0
            score_pi = 1 - pi
            score_lam = 0.0
        else:
            # Zero-Truncated Poisson
            while True:
                z = np.random.poisson(lam)
                if z > 0:
                    y[t] = z
                    break
            # Score da Poisson truncada
            num = lam * np.exp(-lam)
            den = 1 - np.exp(-lam)
            adjustment = num / den
            score_pi = -pi
            score_lam = y[t] - lam * adjustment

        # Atualizações GAS
        f1 = omega1 + alpha1 * score_lam + beta1 * f1
        f2 = omega2 + alpha2 * score_pi + beta2 * f2

        # Clipping dos estados
        f1 = np.clip(f1, *loglam_bounds)
        f2 = np.clip(f2, *logitpi_bounds)

    return y, lambda_t, pi_t
```

```python
n_obs = 300
series_list = []
param_list = []

rng = np.random.default_rng(42)

for i in range(5):
    params_i = {
        'omega1': round(rng.uniform(-0.5, 0.5), 2),
        'alpha1': round(rng.uniform(0.05, 0.25), 2),
        'beta1':  round(rng.uniform(0.5, 0.95), 2),
        'omega2': round(rng.uniform(-0.5, 0.5), 2),
        'alpha2': round(rng.uniform(0.05, 0.25), 2),
        'beta2':  round(rng.uniform(0.5, 0.95), 2),
    }

    y_i, lam_i, pi_i = simulate_ghp(n_obs, params_i, seed=123 + i)
    series_list.append(y_i)
    param_list.append(params_i)
```

```python
fig, axs = plt.subplots(5, 1, figsize=(12, 12), sharex=True)

for i in range(5):
    s = series_list[i]
    p = param_list[i]

    axs[i].plot(s, lw=1.6)

    legend_text = (
        f"ω₁={p['omega1']}, α₁={p['alpha1']}, β₁={p['beta1']}, "
        f"ω₂={p['omega2']}, α₂={p['alpha2']}, β₂={p['beta2']}"
    )

    axs[i].legend([legend_text], loc='upper right')
    axs[i].set_title(f"Série Simulada {i+1}")
    axs[i].grid(True, linestyle='--', alpha=0.6)

axs[-1].set_xlabel("Tempo")
plt.tight_layout()
plt.show()
```

### 1.5 Ajustando modelo GHP

Abaixo, usaremos uma das séries simuladas para encontrar o ajuste minimizando a verossimilhança.......

```python
def ghp_scores(y, lam, pi):
    """
    Retorna os scores reparametrizados (tilde-s) do modelo GHP.
    Score é um VETOR de dimensão 2:
        [score_pi , score_lambda]
    conforme fórmulas (18) e (19)
    """
    if y == 0:
        score_pi = 1 - pi
        score_lam = -lam
    else:
        # Poisson truncada
        trunc_factor = lam * np.exp(-lam) / (1 - np.exp(-lam))
        score_pi = -pi
        score_lam = y - trunc_factor

    return score_pi, score_lam
```

```python
def ghp_filter_paths(
    params, 
    y, 
    loglam_bounds=(-5, 4), 
    logitpi_bounds=(-6, 6)
):
    """
    Constrói os caminhos filtrados de log(λ_t) e logit(π_t),
    com clipping de segurança em ambas equações.
    """
    omega1, alpha1, beta1, omega2, alpha2, beta2 = params

    T = len(y)
    f1 = np.zeros(T)   # log(λ_t)
    f2 = np.zeros(T)   # logit(π_t)

    lambda_t = np.zeros(T)
    pi_t     = np.zeros(T)

    for t in range(T):
        lambda_t[t] = np.exp(f1[t])
        pi_t[t]     = expit(f2[t])

        if t < T - 1:
            score_pi, score_lam = ghp_scores(y[t], lambda_t[t], pi_t[t])

            # Atualização GAS
            f1_t1 = omega1 + alpha1 * score_lam + beta1 * f1[t]
            f2_t1 = omega2 + alpha2 * score_pi  + beta2 * f2[t]

            # Clipping para evitar explosão
            f1[t+1] = np.clip(f1_t1, *loglam_bounds)
            f2[t+1] = np.clip(f2_t1, *logitpi_bounds)

    return lambda_t, pi_t, f1, f2
```

```python
def neg_loglik_ghp(params, y):
    omega1, alpha1, beta1, omega2, alpha2, beta2 = params

    # Restrição de estacionariedade GAS
    if not (0 < beta1 < 1 and 0 < beta2 < 1):
        return np.inf
    if not (0 <= alpha1 <= 1 and 0 <= alpha2 <= 1):
        return np.inf

    lambda_t, pi_t, _, _ = ghp_filter_paths(params, y)

    loglik = 0.0
    for t in range(len(y)):
        lam = lambda_t[t]
        pi0 = pi_t[t]

        if y[t] == 0:
            loglik += np.log(pi0 + (1 - pi0) * np.exp(-lam))
        else:
            pois = y[t] * np.log(lam) - lam - gammaln(y[t] + 1)
            pois_trunc = pois - np.log(1 - np.exp(-lam))
            loglik += np.log(1 - pi0) + pois_trunc

    return -loglik
```

```python
serie_indice = 0
y_obs = series_list[serie_indice]
n = len(y_obs)
parametros_originais = param_list[serie_indice]

# Chute inicial sensato
x0 = np.array([0.1, 0.1, 0.9, 0.1, 0.1, 0.9])

bounds = [
    (-2, 2),   # omega1
    (0, 1),    # alpha1
    (0, 1),    # beta1
    (-2, 2),   # omega2
    (0, 1),    # alpha2
    (0, 1)     # beta2
]

result = minimize(
    neg_loglik_ghp,
    x0,
    args=(y_obs,),
    method="L-BFGS-B",
    bounds=bounds
)
```

### Modelo GAS Hurdle Poisson — Valor Ajustado $\hat{y}_t$

O modelo **GAS Hurdle Poisson (GHP)** combina dois mecanismos:

1. Uma **barreira probabilística** para geração de zeros  
2. Uma **distribuição de Poisson truncada em zero** para contagens positivas

A distribuição condicional de $Y_t$ é dada por:

$$
Y_t = 
\begin{cases}
0, & \text{com probabilidade } \pi_t, \\
Z_t, & \text{com probabilidade } 1 - \pi_t,\quad Z_t \sim \text{Poisson}_+( \lambda_t),
\end{cases}
$$

onde $\text{Poisson}_+(\lambda)$ é a Poisson truncada em zero.

---

### 1. Probabilidade de Zero

A probabilidade $\pi_t$ é modelada dinamicamente na escala logito:

$$
\text{logit}(\pi_t) = \eta_t = \omega_2 + \beta_2 \eta_{t-1} + \alpha_2 s_{\pi, t-1}
$$

com o score:

$$
s_{\pi, t-1} =
\begin{cases}
1 - \pi_{t-1}, & \text{se } y_{t-1} = 0 \\
- \pi_{t-1}, & \text{se } y_{t-1} > 0
\end{cases}
$$

e

$$
\pi_t = \frac{1}{1 + e^{-\eta_t}}.
$$

---

### 2. Intensidade da Poisson Truncada

A intensidade $\lambda_t$ também é modelada dinamicamente via equação recursiva:

$$
\log \lambda_t = f_t = \omega_1 + \beta_1 f_{t-1} + \alpha_1 s_{\lambda, t-1}
$$

com o score:

$$
s_{\lambda, t-1} =
\begin{cases}
0, & \text{se } y_{t-1} = 0 \\
y_{t-1} - \lambda_{t-1} \cdot \dfrac{e^{-\lambda_{t-1}}}{1 - e^{-\lambda_{t-1}}}, & \text{se } y_{t-1} > 0
\end{cases}
$$

e

$$
\lambda_t = \exp(f_t).
$$

---

### 3. Média da Poisson Truncada

Se $Z_t \sim \text{Poisson}(\lambda_t)$ condicionado a $Z_t > 0$, a média condicional é:

$$
\mu_t = \mathbb{E}[Z_t \mid Z_t > 0] = \frac{\lambda_t}{1 - e^{-\lambda_t}}.
$$

---

### 4. Valor Ajustado $\hat{y}_t$

A esperança condicional total é:

$$
\mathbb{E}[Y_t \mid \mathcal{F}_{t-1}] 
= \pi_t \cdot 0 + (1 - \pi_t) \cdot \mu_t
= (1 - \pi_t) \cdot \frac{\lambda_t}{1 - e^{-\lambda_t}}.
$$

Portanto:

$$
\boxed{
\hat{y}_t = (1 - \pi_t) \cdot \frac{\lambda_t}{1 - e^{-\lambda_t}}
}
$$

Essa é a curva ajustada $\hat{y}_t$, que representa a média condicional de $Y_t$ e serve como base para previsões e métricas de erro.

---

### 5. Interpretação

- $\lambda_t$ regula a intensidade esperada das contagens positivas.
- $\pi_t$ determina a chance de ocorrência de zeros estruturais.
- A média ajustada $\hat{y}_t$ reflete o efeito conjunto da barreira de zeros e da Poisson truncada.
- $\hat{y}_t$ é útil para comparação com os dados observados, avaliação de erro e previsão out-of-sample.

## Diagnóstico do GHP

```python
def diagnostics_ghp(y, lambda_t, pi_t, title='Diagnóstico do Modelo GHP'):
    T = len(y)
    pit = np.zeros(T)
    quant_resid = np.zeros(T)
    pearson_resid = np.zeros(T)

    for t in range(T):
        y_t = y[t]
        lam = lambda_t[t]
        pi = pi_t[t]
        p0 = np.exp(-lam)
        p0 = np.clip(p0, 1e-12, 1 - 1e-12)  # segurança

        if y_t == 0:
            # PIT
            pit[t] = np.random.uniform(0, pi)
            # Resíduo quantílico
            quant_resid[t] = norm.ppf(pit[t])
            # Pearson
            mu = (1 - pi) * lam / (1 - p0)
            var = (1 - pi) * (lam * (1 + lam * p0 / (1 - p0)) / (1 - p0)) - mu**2
            pearson_resid[t] = (0 - mu) / np.sqrt(var)

        else:
            # CDF da Poisson truncada
            def F_ztrunc(k):
                return (poisson.cdf(k, lam) - p0) / (1 - p0)

            F_lo = F_ztrunc(y_t - 1) if y_t > 1 else 0
            F_hi = F_ztrunc(y_t)
            u = np.random.uniform()
            pit[t] = pi + (1 - pi) * (F_lo + u * (F_hi - F_lo))
            quant_resid[t] = norm.ppf(pit[t])

            # Esperança e variância condicional
            mu = (1 - pi) * lam / (1 - p0)
            var_z = lam * (1 + lam * p0 / (1 - p0))  # var da Poisson truncada
            var = (1 - pi) * (var_z + pi * (lam / (1 - p0))**2)
            pearson_resid[t] = (y_t - mu) / np.sqrt(var)

    # Gráficos
    fig, axs = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle(f'{title} – PIT e Resíduos', fontsize=16)

    sns.histplot(pit, bins=20, kde=False, ax=axs[0, 0], color='skyblue')
    axs[0, 0].set_title('Histograma do PIT')
    axs[0, 0].axhline(T / 20, color='red', linestyle='--')

    sns.histplot(quant_resid, bins=30, kde=True, ax=axs[0, 1])
    axs[0, 1].set_title('Resíduos Quantílicos')

    probplot(quant_resid, dist="norm", plot=axs[1, 0])
    axs[1, 0].set_title('QQ-Plot do Resíduo Quantílico')

    axs[1, 1].scatter(range(T), pearson_resid, alpha=0.7)
    axs[1, 1].set_title('Resíduos de Pearson')

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

    return pit, quant_resid, pearson_resid
```

## Previsão $k$ passos à frente

```python
# Função de previsão com simulação
def forecast_ghp_k_steps_sim(
    k, params, f1_T, f2_T, n_sim=500, seed=None
):
    if seed is not None:
        np.random.seed(seed)

    omega1, alpha1, beta1 = params['omega1'], params['alpha1'], params['beta1']
    omega2, alpha2, beta2 = params['omega2'], params['alpha2'], params['beta2']

    sim_y = np.zeros((n_sim, k))
    sim_lambda = np.zeros((n_sim, k))
    sim_pi = np.zeros((n_sim, k))

    for i in range(n_sim):
        f1 = f1_T
        f2 = f2_T

        for h in range(k):
            # Score esperado nulo
            f1 = omega1 + beta1 * f1
            f2 = omega2 + beta2 * f2

            lambda_t = np.exp(f1)
            pi_t = expit(f2)

            sim_lambda[i, h] = lambda_t
            sim_pi[i, h] = pi_t

            # Geração de y_t
            if np.random.rand() < pi_t:
                y_t = 0
            else:
                while True:
                    z = np.random.poisson(lambda_t)
                    if z > 0:
                        y_t = z
                        break
            sim_y[i, h] = y_t

    # Intervalos de confiança e previsão média
    yhat_mean = sim_y.mean(axis=0)
    yhat_lb = np.percentile(sim_y, 2.5, axis=0)
    yhat_ub = np.percentile(sim_y, 97.5, axis=0)

    return yhat_mean, yhat_lb, yhat_ub, sim_y
```

```python
# Gerar série simulada
y_obs, lambda_obs, pi_obs = simulate_ghp(300, parametros_originais, seed=123)

# Forecast k-passos
f1_T = np.log(lambda_obs[-1])
f2_T = logit(pi_obs[-1])
```

```python
params_hat = {
    'omega1': θ_hat[0],
    'alpha1': θ_hat[1],
    'beta1': θ_hat[2],
    'omega2': θ_hat[3],
    'alpha2': θ_hat[4],
    'beta2': θ_hat[5],
}

k = 10
yhat_mean, yhat_lb, yhat_ub, sim_y = forecast_ghp_k_steps_sim(
    k, params_hat, f1_T, f2_T, n_sim=500, seed=42
)
```

```python
def simulate_forecast_ghp(k, y_obs, params, n_sim=1000, seed=None):
    """
    Previsão k-passos à frente com simulação para ICs.

    Parâmetros:
    - k: horizonte de previsão
    - y_obs: série observada
    - params: vetor de parâmetros estimados [omega, alpha, beta, rho]
    - n_sim: número de trajetórias simuladas
    - seed: semente para reprodutibilidade

    Retorna:
    - y_sim: array (n_sim, k) com trajetórias simuladas
    - y_mean: previsão média por passo
    - y_ci_low: percentil inferior (2.5%)
    - y_ci_high: percentil superior (97.5%)
    """
    if seed is not None:
        np.random.seed(seed)

    omega, alpha, beta, rho = params
    f_last = omega

    # Recalcular f_t com dados observados
    for y in y_obs:
        lam = np.exp(f_last)
        score = y - lam
        f_last = omega + alpha * score + beta * f_last
        f_last = np.clip(f_last, -5, 4)

    y_sim = np.zeros((n_sim, k))

    for s in range(n_sim):
        f = f_last
        for h in range(k):
            lam = np.exp(f)
            lam = np.clip(lam, 1e-6, 50)
            pi = 1 / (1 + np.exp(-rho))

            u = np.random.rand()
            if u < pi:
                y_sim[s, h] = 0
            else:
                while True:
                    z = np.random.poisson(lam)
                    if z > 0:
                        y_sim[s, h] = z
                        break

            score = y_sim[s, h] - lam
            f = omega + alpha * score + beta * f
            f = np.clip(f, -5, 4)

    # Estatísticas
    y_mean = np.mean(y_sim, axis=0)
    y_ci_low = np.percentile(y_sim, 2.5, axis=0)
    y_ci_high = np.percentile(y_sim, 97.5, axis=0)

    return y_sim, y_mean, y_ci_low, y_ci_high
```

```python
# Executar previsão com IC
k = 20
n_sim = 2_000
y_sim, y_mean, y_ci_low, y_ci_high = simulate_forecast_ghp(k, y_obs, result.x, n_sim=n_sim, seed=42)
```

# 2. GAS Hurdle Negative Binomial (GHNB)

O modelo hurdle binomial negativa (HNB) segue a mesma lógica do HP, porém a parte positiva é amostrada de uma **Binomial Negativa truncada em zero**.

A função de massa é:

$$
p(y_t \mid \mu_{t|t-1}, \alpha, \pi_{t|t-1}) =
\begin{cases}
\pi_{t|t-1}, & y_t = 0, \\[8pt]
(1 - \pi_{t|t-1})
\dfrac{
\Gamma(y_t + \alpha)
}{
\Gamma(y_t + 1)\Gamma(\alpha)
}
\left(
\frac{1}{1 + \frac{\mu_{t|t-1}}{\alpha}}
\right)^{\alpha}
\left(
1 - 
\frac{1}{
1 + \frac{\mu_{t|t-1}}{\alpha}
}
\right)^{y_t}
\dfrac{1}{
1 - \left(1 + \frac{\mu_{t|t-1}}{\alpha}\right)^{-\alpha}
},
& y_t > 0.
\end{cases}
\tag{20}
$$

com  
- $\mu_{t|t-1} > 0$,  
- $\alpha > 0$,  
- $\pi_{t|t-1} \in (0,1)$.

## 2.1 Média e Variância

$$
\mathbb{E}_{t-1}[y_t]
=
\frac{\mu_{t|t-1}(1 - \pi_{t|t-1})}{
1 - \left(1 + \frac{\mu_{t|t-1}}{\alpha}\right)^{-\alpha}}
$$

$$
\operatorname{Var}_{t-1}[y_t]
=
\pi_{t|t-1}(1 - \pi_{t|t-1})
\left(
\frac{\mu_{t|t-1}}{
1 - \left(1 + \frac{\mu_{t|t-1}}{\alpha}\right)^{-\alpha}
}
\right)^2
+
(1 - \pi_{t|t-1})
\left[
\dfrac{
\mu_{t|t-1}\left(1 + \frac{\mu_{t|t-1}}{\alpha}\right)
}{
1 - \left(1 + \frac{\mu_{t|t-1}}{\alpha}\right)^{-\alpha}}
-
\left(1 + \frac{\mu_{t|t-1}}{\alpha}\right)^{-\alpha}
\left(
\frac{\mu_{t|t-1}}{
1 -
\left(1 + \frac{\mu_{t|t-1}}{\alpha}\right)^{-\alpha}}
\right)^2
\right]
\tag{21}
$$

## 2.2 Score Reparametrizado

Zero:

$$
\tilde{s}_{y_t=0}=
\begin{bmatrix}
1-\pi_{t|t-1}\\
0
\end{bmatrix}
\tag{22}
$$

Positivo:
$$
\tilde{s}_{y_t>0}=
\begin{bmatrix}
-\pi_{t|t-1}\\[6pt]
\dfrac{
y_t - 
\mu_{t|t-1}\left(1 + \frac{\mu_{t|t-1}}{\alpha}\right)^{-(\alpha+1)}
}{
1 - \left(1 + \frac{\mu_{t|t-1}}{\alpha}\right)^{-\alpha}
}
\end{bmatrix}
\tag{23}
$$

Assim como no HP, o mecanismo de atualização gera previsões k‑passos‑à‑frente de forma totalmente probabilística e coerente.

```python
# Função de simulação GAS-HNB
def simulate_ghnb(
    n_obs, 
    params, 
    seed=None, 
    lambda_max=50, 
    loglam_bounds=(-5, 4), 
    logitpi_bounds=(-6, 6)
):
    
    """
    Simula uma série temporal do modelo GAS Hurdle Negative Binomial (GHNB)
    com dinâmica para log(μ_t) e logit(π_t), conforme especificação teórica.

    Parâmetros:
    - n_obs: número de observações
    - params: dicionário com chaves:
        - omega1, alpha1, beta1: dinâmica de log(μ_t)
        - omega2, alpha2, beta2: dinâmica de logit(π_t)
        - sigma: desvio padrão do ruído (não usado aqui)
        - alpha: dispersão da NB
    - seed: semente aleatória

    Retorna:
    - y: série simulada
    - mu_t: sequência de médias da NB
    - pi_t: sequência de probabilidades de zero
    """

    if seed is not None:
        np.random.seed(seed)

    # Parâmetros da dinâmica para log(μ_t)
    omega1 = params['omega1']
    alpha1 = params['alpha1']
    beta1  = params['beta1']

    # Parâmetros da dinâmica para logit(π_t)
    omega2 = params['omega2']
    alpha2 = params['alpha2']
    beta2  = params['beta2']

    # Dispersão da NB (α > 0)
    alpha_disp = params['alpha']
    
    y = np.zeros(n_obs, dtype=int)
    mu_seq = np.zeros(n_obs)
    pi_seq = np.zeros(n_obs)

    # Inicialização dos estados latentes
    f1 = omega1  # log(μ_1)
    f2 = omega2  # logit(π_1)

    for t in range(n_obs):
        # Transformações
        mu_t = np.clip(np.exp(f1), 1e-6, lambda_max)  # μ_t > 0
        eta_pi = np.clip(f2, logitpi_bounds[0], logitpi_bounds[1])
        pi_t = 1 / (1 + np.exp(-eta_pi))  # logit⁻¹(f2)

        mu_seq[t] = mu_t
        pi_seq[t] = pi_t

        # Geração de y_t
        u = np.random.rand()
        if u < pi_t:
            y[t] = 0
        else:
            # Gera da NB truncada em zero
            while True:
                p_nb = alpha_disp / (alpha_disp + mu_t)
                z = np.random.negative_binomial(alpha_disp, p_nb)
                if z > 0:
                    y[t] = z
                    break

        # Score reparametrizado
        one_plus = 1 + mu_t / alpha_disp
        one_plus_pow_alpha = one_plus ** (-alpha_disp)

        if y[t] == 0:
            score1 = 1 - pi_t      # score para π
            score2 = 0             # score para μ
        else:
            numer = y[t] - mu_t * one_plus ** (-alpha_disp - 1)
            denom = 1 - one_plus_pow_alpha
            score1 = -pi_t
            score2 = numer / denom

        # Atualização dos estados
        f1 = omega1 + alpha1 * score2 + beta1 * f1
        f2 = omega2 + alpha2 * score1 + beta2 * f2

        # Clipping para estabilidade
        f1 = np.clip(f1, loglam_bounds[0], loglam_bounds[1])
        f2 = np.clip(f2, logitpi_bounds[0], logitpi_bounds[1])

    return y, mu_seq, pi_seq
```

```python
n_obs = 200
series_list = []
param_list = []

rng = np.random.default_rng(42)

for i in range(5):
    print(f"Simulando série {i+1}")

    params_i = {
        # Dinâmica para log(mu_t)
        'omega1': rng.uniform(-0.5, 0.5),
        'alpha1': rng.uniform(0.05, 0.25),
        'beta1':  rng.uniform(0.5, 0.95),

        # Dinâmica para logit(pi_t)
        'omega2': rng.uniform(-1.0, 1.0),
        'alpha2': rng.uniform(0.05, 0.25),
        'beta2':  rng.uniform(0.5, 0.95),

        # Dispersão da NB
        'alpha': rng.uniform(2, 6)
    }

    y_i, mu_i, pi_i = simulate_ghnb(n_obs, params_i, seed=123+i)
    series_list.append((y_i, mu_i, pi_i))
    param_list.append(params_i)
```

```python
def ghnb_scores(y, mu, pi, alpha):
    """
    Retorna os scores reparametrizados (tilde-s) do modelo GHNB.
    Score é um VETOR de dimensão 2: [score_pi , score_mu]
    conforme fórmulas (22) e (23).
    """
    if y == 0:
        score_pi = 1 - pi
        score_mu = 0.0
    else:
        A = 1 + mu / alpha
        denom = 1 - A ** (-alpha)
        numer = y - mu * A ** (-(alpha + 1))
        score_pi = -pi
        score_mu = numer / denom

    return score_pi, score_mu
```

```python
def ghnb_filter_paths(params, y, logmu_bounds=(-5, 4), logitpi_bounds=(-6, 6)):
    """
    Constrói os caminhos filtrados de log(μ_t) e logit(π_t),
    e retorna também os scores reparametrizados.
    """
    omega1, alpha1, beta1, omega2, alpha2, beta2, log_alpha = params
    alpha_nb = np.exp(log_alpha)

    T = len(y)
    f1 = np.zeros(T)   # log(μ_t)
    f2 = np.zeros(T)   # logit(π_t)

    mu_t = np.zeros(T)
    pi_t = np.zeros(T)
    score_pi = np.zeros(T)
    score_mu = np.zeros(T)

    for t in range(T):
        mu_t[t] = np.exp(f1[t])
        pi_t[t] = expit(f2[t])

        if t < T - 1:
            score_pi[t], score_mu[t] = ghnb_scores(y[t], mu_t[t], pi_t[t], alpha_nb)

            # Atualização GAS
            f1_t1 = omega1 + alpha1 * score_mu[t] + beta1 * f1[t]
            f2_t1 = omega2 + alpha2 * score_pi[t] + beta2 * f2[t]

            f1[t + 1] = np.clip(f1_t1, *logmu_bounds)
            f2[t + 1] = np.clip(f2_t1, *logitpi_bounds)

    return mu_t, pi_t, f1, f2, score_pi, score_mu
```

```python
def neg_loglik_ghnb(params, y):
    """
    Log-verossimilhança negativa do modelo GAS Hurdle Negative Binomial (GHNB),
    com atualização das componentes log(μ_t) e logit(π_t).
    """
    omega1, alpha1, beta1, omega2, alpha2, beta2, log_alpha = params
    alpha_nb = np.exp(log_alpha)

    # Restrições básicas de estacionariedade e estabilidade
    if not (0 < beta1 < 1 and 0 < beta2 < 1):
        return np.inf
    if not (0 <= alpha1 <= 1 and 0 <= alpha2 <= 1):
        return np.inf

    mu_t, pi_t, _, _, _, _ = ghnb_filter_paths(params, y)

    loglik = 0.0
    for t in range(len(y)):
        mu = mu_t[t]
        pi = pi_t[t]
        A = 1 + mu / alpha_nb
        r = alpha_nb

        if y[t] == 0:
            loglik += np.log(pi)
        else:
            # Parte da NB
            log_p_nb = (
                gammaln(y[t] + r) - gammaln(r) - gammaln(y[t] + 1)
                + r * np.log(r / (r + mu))
                + y[t] * np.log(mu / (r + mu))
            )
            # Truncação em zero
            log_trunc = -np.log(1 - (r / (r + mu)) ** r)
            log_p_nb += log_trunc

            loglik += np.log(1 - pi) + log_p_nb

    return -loglik
```

```python
indice_serie = 4
y_obs_nb = series_list[indice_serie][0]

init_params = [
    0.1, 0.1, 0.8,    # ω1, α1, β1
    0.1, 0.1, 0.8,    # ω2, α2, β2
    np.log(3.0)       # log(alpha_disp)
]

result_nb = minimize(
    neg_loglik_ghnb,
    init_params,
    args=(y_obs_nb,),
    method='L-BFGS-B'
)

omega1_hat, alpha1_hat, beta1_hat, omega2_hat, alpha2_hat, beta2_hat, log_alpha_hat = result_nb.x
alpha_disp_hat = np.exp(log_alpha_hat)
```

```python
def diagnostics_ghnb(y_obs, lambda_hat, pi_hat, r_hat, title='Diagnóstico GHNB'):
    """
    Diagnósticos do modelo GAS Hurdle Negative Binomial
    - PIT com hurdle
    - Resíduo quantílico (QQ plot)
    """
    n = len(y_obs)
    pit = np.zeros(n)
    quant_resid = np.zeros(n)

    for t in range(n):
        y = y_obs[t]
        λ = lambda_hat[t]
        π = pi_hat[t]
        r = r_hat

        # p = r / (r + λ)
        p_nb = r / (r + λ)

        if y == 0:
            pit[t] = uniform.rvs(0, π)
        else:
            # CDF truncada da NB em zero
            def cdf_ztnb(k):
                numer = nbinom.cdf(k, r, p_nb) - nbinom.pmf(0, r, p_nb)
                denom = 1 - nbinom.pmf(0, r, p_nb)
                return numer / denom

            pit_nb = cdf_ztnb(y - 1) + np.random.uniform() * (cdf_ztnb(y) - cdf_ztnb(y - 1))
            pit[t] = π + (1 - π) * pit_nb

    # Resíduo quantílico
    quant_resid = norm.ppf(pit)

    # Plots
    fig, axs = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle(title, fontsize=16)

    # PIT
    sns.histplot(pit, bins=20, kde=False, stat='density', ax=axs[0], color='steelblue')
    axs[0].plot([0, 1], [1, 1], 'r--')
    axs[0].set_title("Histograma do PIT")
    axs[0].set_xlabel("PIT")
    axs[0].set_ylabel("Densidade")

    # QQ-plot dos resíduos quantílicos
    probplot(quant_resid, dist="norm", plot=axs[1])
    axs[1].set_title("QQ-Plot dos Resíduos Quantílicos")

    plt.tight_layout()
    plt.show()

    return pit, quant_resid
```

## Previsao $k$ passos a frente

```python
def forecast_ghnb(
    f1_last, 
    f2_last, 
    params, 
    k=10, 
    n_sim=5000,
    loglam_bounds=(-6, 6), 
    logitpi_bounds=(-6, 6)
):
    """
    Previsão k-passos à frente por simulação Monte Carlo para GHNB.
    """

    omega1, alpha1, beta1 = params["omega1"], params["alpha1"], params["beta1"]
    omega2, alpha2, beta2 = params["omega2"], params["alpha2"], params["beta2"]
    alpha_disp = params["alpha"]

    forecasts = np.zeros((n_sim, k))

    for s in range(n_sim):

        f1 = f1_last
        f2 = f2_last

        for h in range(k):

            mu = np.exp(f1)
            mu = np.clip(mu, 1e-6, 50)

            pi = 1 / (1 + np.exp(-f2))
            pi = np.clip(pi, 1e-6, 1-1e-6)

            # Simulação hurdle NB
            u = np.random.rand()
            if u < pi:
                y_sim = 0
            else:
                while True:
                    p_nb = alpha_disp / (alpha_disp + mu)
                    y_sim = np.random.negative_binomial(alpha_disp, p_nb)
                    if y_sim > 0:
                        break

            forecasts[s, h] = y_sim

            # Score reparametrizado
            if y_sim == 0:
                score1 = 1 - pi
                score2 = 0
            else:
                one_plus = 1 + mu / alpha_disp
                score1 = -pi
                score2 = (y_sim - mu * one_plus**(-(alpha_disp+1))) / (1 - one_plus**(-alpha_disp))

            # GAS update
            f1 = omega1 + alpha1 * score2 + beta1 * f1
            f2 = omega2 + alpha2 * score1 + beta2 * f2

            f1 = np.clip(f1, loglam_bounds[0], loglam_bounds[1])
            f2 = np.clip(f2, logitpi_bounds[0], logitpi_bounds[1])

    mean_forecast = forecasts.mean(axis=0)
    ic = np.percentile(forecasts, [2.5, 97.5], axis=0)

    return mean_forecast, ic
```

```python
def forecast_ghnb(f_last, params, k=10, n_sim=5000, loglam_bounds=(-5, 4)):
    """
    Previsão k-passos à frente para o modelo GAS Hurdle NB via simulação.

    Parâmetros:
    - f_last: último valor de f_t (log-lambda)
    - params: dicionário com parâmetros estimados: omega, alpha, beta, rho, r
    - k: horizonte de previsão
    - n_sim: número de simulações para IC
    - loglam_bounds: limites para clipping de f

    Retorna:
    - forecast_mean: valor médio da previsão para cada passo
    - forecast_ic: intervalo de confiança 95% (inferior e superior)
    """
    omega = params['omega']
    alpha = params['alpha']
    beta  = params['beta']
    rho   = params['rho']
    r     = params['r']

    forecasts = np.zeros((n_sim, k))

    for i in range(n_sim):
        f_t = f_last
        for h in range(k):
            lambda_t = np.exp(f_t)
            lambda_t = np.clip(lambda_t, 1e-6, 50)

            p_zero = 1 / (1 + np.exp(-rho))  # hurdle
            p_nb   = r / (r + lambda_t)

            # Simular hurdle NB
            u = np.random.rand()
            if u < p_zero:
                y_sim = 0
            else:
                # Zero-truncated NB
                while True:
                    y_sim = np.random.negative_binomial(r, p_nb)
                    if y_sim > 0:
                        break

            forecasts[i, h] = y_sim

            # GAS update (usando score de NB)
            score = y_sim - lambda_t
            f_t = omega + alpha * score + beta * f_t
            f_t = np.clip(f_t, loglam_bounds[0], loglam_bounds[1])

    # Estatísticas
    forecast_mean = np.mean(forecasts, axis=0)
    forecast_ic = np.percentile(forecasts, [2.5, 97.5], axis=0)

    return forecast_mean, forecast_ic
```

```python
# === codigo refatorado ===
from typing import Tuple, Dict
import numpy as np
from scipy.special import expit, gammaln

# ----------------------------
# UTILS
# ----------------------------
def sigmoid(x):
    return expit(x)

def clip_logit(x, bounds=(-6, 6)):
    return np.clip(x, *bounds)

def clip_exp(x, bounds=(-5, 4)):
    return np.clip(np.exp(x), np.exp(bounds[0]), np.exp(bounds[1]))

# ----------------------------
# SCORES
# ----------------------------
def score_ghp(y, lam, pi):
    if y == 0:
        return 1 - pi, -lam
    else:
        adjustment = lam * np.exp(-lam) / (1 - np.exp(-lam))
        return -pi, y - adjustment

def score_ghnb(y, mu, pi, alpha):
    if y == 0:
        return 1 - pi, 0.0
    else:
        A = 1 + mu / alpha
        denom = 1 - A ** (-alpha)
        numer = y - mu * A ** (-(alpha + 1))
        return -pi, numer / denom

# ----------------------------
# SIMULAÇÃO
# ----------------------------
def simulate_gas_model(
    n_obs: int,
    params: dict,
    model: str = 'ghp', 
    seed=None
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    if seed is not None:
        np.random.seed(seed)

    omega1, alpha1, beta1 = params['omega1'], params['alpha1'], params['beta1']
    omega2, alpha2, beta2 = params['omega2'], params['alpha2'], params['beta2']
    alpha = params.get('alpha', None)  # apenas para GHNB

    f1, f2 = omega1, omega2
    y, m1_seq, pi_seq = np.zeros(n_obs, dtype=int), np.zeros(n_obs), np.zeros(n_obs)

    for t in range(n_obs):
        m1 = np.clip(np.exp(f1), 1e-6, 50)  # lambda (GHP) ou mu (GHNB)
        pi = sigmoid(np.clip(f2, -6, 6))
        m1_seq[t], pi_seq[t] = m1, pi

        u = np.random.rand()
        if u < pi:
            y[t] = 0
            score_pi, score_m1 = 1 - pi, 0.0
        else:
            if model == 'ghp':
                while True:
                    z = np.random.poisson(m1)
                    if z > 0:
                        y[t] = z
                        break
                score_pi, score_m1 = score_ghp(y[t], m1, pi)

            elif model == 'ghnb':
                while True:
                    p_nb = alpha / (alpha + m1)
                    z = np.random.negative_binomial(alpha, p_nb)
                    if z > 0:
                        y[t] = z
                        break
                score_pi, score_m1 = score_ghnb(y[t], m1, pi, alpha)

        f1 = omega1 + alpha1 * score_m1 + beta1 * f1
        f2 = omega2 + alpha2 * score_pi + beta2 * f2
        f1 = np.clip(f1, -5, 4)
        f2 = np.clip(f2, -6, 6)

    return y, m1_seq, pi_seq

# ----------------------------
# FILTRAGEM
# ----------------------------
def filter_gas_model(
    y: np.ndarray, 
    params: dict, 
    model: str = 'ghp'
) -> Tuple[np.ndarray, np.ndarray, np.ndarray, np.ndarray]:
    omega1, alpha1, beta1 = params['omega1'], params['alpha1'], params['beta1']
    omega2, alpha2, beta2 = params['omega2'], params['alpha2'], params['beta2']
    alpha = params.get('alpha', None)

    T = len(y)
    f1 = np.zeros(T)
    f2 = np.zeros(T)
    m1_seq = np.zeros(T)
    pi_seq = np.zeros(T)

    for t in range(T):
        m1_seq[t] = np.exp(f1[t])
        pi_seq[t] = expit(f2[t])

        if t < T - 1:
            if model == 'ghp':
                score_pi, score_m1 = score_ghp(y[t], m1_seq[t], pi_seq[t])
            elif model == 'ghnb':
                score_pi, score_m1 = score_ghnb(y[t], m1_seq[t], pi_seq[t], alpha)
            else:
                raise ValueError("Modelo desconhecido")

            f1[t+1] = np.clip(omega1 + alpha1 * score_m1 + beta1 * f1[t], -5, 4)
            f2[t+1] = np.clip(omega2 + alpha2 * score_pi + beta2 * f2[t], -6, 6)

    return m1_seq, pi_seq, f1, f2
```

```python
# --------------------------------------------------
# Scores Reparametrizados
# --------------------------------------------------

def score_ghp(y, lam, pi):
    if y == 0:
        return 1 - pi, -lam
    else:
        trunc = lam * np.exp(-lam) / (1 - np.exp(-lam))
        return -pi, y - trunc

def score_ghnb(y, mu, pi, alpha):
    if y == 0:
        return 1 - pi, 0.0
    else:
        denom = 1 - (1 + mu / alpha) ** (-alpha)
        part = mu * (1 + mu / alpha) ** (-(alpha + 1))
        score_mu = (y - part) / denom
        return -pi, score_mu

# --------------------------------------------------
# Simulador Geral GAS Hurdle
# --------------------------------------------------

def simulate_gas_model(n_obs, params, model='ghp', seed=None, mu_max=50, logmu_bounds=(-5, 4)):
    if seed is not None:
        np.random.seed(seed)

    omega1 = params['omega1']
    alpha1 = params['alpha1']
    beta1  = params['beta1']
    omega2 = params['omega2']
    alpha2 = params['alpha2']
    beta2  = params['beta2']
    alpha_nb = np.exp(params['log_alpha']) if model == 'ghnb' else None

    y = np.zeros(n_obs, dtype=int)
    mu = np.zeros(n_obs)
    pi = np.zeros(n_obs)
    f1 = omega1
    f2 = omega2

    for t in range(n_obs):
        mu_t = np.clip(np.exp(f1), 1e-6, mu_max)
        pi_t = expit(f2)
        mu[t] = mu_t
        pi[t] = np.clip(pi_t, 1e-6, 1 - 1e-6)

        u = np.random.rand()
        if u < pi[t]:
            y[t] = 0
        else:
            if model == 'ghp':
                while True:
                    z = np.random.poisson(mu_t)
                    if z > 0:
                        y[t] = z
                        break
            elif model == 'ghnb':
                while True:
                    p_nb = alpha_nb / (alpha_nb + mu_t)
                    z = np.random.negative_binomial(alpha_nb, p_nb)
                    if z > 0:
                        y[t] = z
                        break

        if model == 'ghp':
            score_pi, score_mu = score_ghp(y[t], mu_t, pi[t])
        elif model == 'ghnb':
            score_pi, score_mu = score_ghnb(y[t], mu_t, pi[t], alpha_nb)

        f1 = np.clip(omega1 + alpha1 * score_mu + beta1 * f1, *logmu_bounds)
        f2 = np.clip(omega2 + alpha2 * score_pi + beta2 * f2, -6, 6)

    return y, mu, pi

# --------------------------------------------------
# Filtro Geral GAS Hurdle (caminhos filtrados)
# --------------------------------------------------

def filter_gas_model(params, y, model='ghp', logmu_bounds=(-5, 4), logitpi_bounds=(-6, 6)):
    omega1, alpha1, beta1, omega2, alpha2, beta2, log_alpha = params
    alpha_nb = np.exp(log_alpha) if model == 'ghnb' else None

    T = len(y)
    f1 = np.zeros(T)
    f2 = np.zeros(T)
    mu_t = np.zeros(T)
    pi_t = np.zeros(T)
    score_mu = np.zeros(T)
    score_pi = np.zeros(T)

    for t in range(T):
        mu_t[t] = np.exp(f1[t])
        pi_t[t] = expit(f2[t])

        if t < T - 1:
            if model == 'ghp':
                score_pi[t], score_mu[t] = score_ghp(y[t], mu_t[t], pi_t[t])
            elif model == 'ghnb':
                score_pi[t], score_mu[t] = score_ghnb(y[t], mu_t[t], pi_t[t], alpha_nb)

            f1[t+1] = np.clip(omega1 + alpha1 * score_mu[t] + beta1 * f1[t], *logmu_bounds)
            f2[t+1] = np.clip(omega2 + alpha2 * score_pi[t] + beta2 * f2[t], *logitpi_bounds)

    return mu_t, pi_t, f1, f2, score_mu, score_pi

# --------------------------------------------------
# Log-Verossimilhança Geral (GHP / GHNB)
# --------------------------------------------------

def loglike_gas_model(params, y, model='ghp'):
    omega1, alpha1, beta1, omega2, alpha2, beta2, log_alpha = params
    alpha_nb = np.exp(log_alpha) if model == 'ghnb' else None

    if not (0 < beta1 < 1 and 0 < beta2 < 1):
        return np.inf

    mu_t, pi_t, _, _, _, _ = filter_gas_model(params, y, model)

    loglike = 0.0
    for t in range(len(y)):
        mu = mu_t[t]
        pi0 = pi_t[t]

        if y[t] == 0:
            if model == 'ghp':
                loglike += np.log(pi0 + (1 - pi0) * np.exp(-mu))
            elif model == 'ghnb':
                trunc = 1 - (1 + mu / alpha_nb) ** (-alpha_nb)
                loglike += np.log(pi0 + (1 - pi0) * (1 - trunc))
        else:
            if model == 'ghp':
                pois = y[t] * np.log(mu) - mu - gammaln(y[t] + 1)
                pois_trunc = pois - np.log(1 - np.exp(-mu))
                loglike += np.log(1 - pi0) + pois_trunc
            elif model == 'ghnb':
                logpmf = (
                    gammaln(y[t] + alpha_nb) - gammaln(y[t] + 1) - gammaln(alpha_nb)
                    + alpha_nb * np.log(alpha_nb / (alpha_nb + mu))
                    + y[t] * np.log(mu / (alpha_nb + mu))
                )
                trunc_corr = np.log(1 - (alpha_nb / (alpha_nb + mu)) ** alpha_nb)
                loglike += np.log(1 - pi0) + logpmf - trunc_corr

    return -loglike
```

```python
def forecast_gas_model(f1_last, f2_last, params, k=10, n_sim=5000, model='ghp',
                       logmu_bounds=(-5, 4), logitpi_bounds=(-6, 6)):
    """
    Simula previsões k-passos à frente para o modelo GAS (GHP ou GHNB).

    Parâmetros:
    - f1_last, f2_last: últimos valores dos estados latentes (log-μ, logit-π)
    - params: parâmetros estimados
    - k: horizonte de previsão
    - n_sim: número de simulações para intervalos de confiança
    - model: 'ghp' ou 'ghnb'
    - bounds: limites para clipping dos estados latentes

    Retorna:
    - forecast_mean: média preditiva k-passos
    - forecast_ic: intervalo de confiança 95%
    """
    if model == 'ghp':
        omega1, alpha1, beta1, omega2, alpha2, beta2 = params
        r = None
    elif model == 'ghnb':
        omega1, alpha1, beta1, omega2, alpha2, beta2, log_alpha = params
        r = np.exp(log_alpha)
    else:
        raise ValueError("Modelo inválido. Use 'ghp' ou 'ghnb'.")

    forecasts = np.zeros((n_sim, k))

    for i in range(n_sim):
        f1_t = f1_last
        f2_t = f2_last

        for h in range(k):
            mu_t = np.exp(f1_t)
            pi_t = expit(f2_t)

            u = np.random.rand()
            if u < pi_t:
                y_sim = 0
            else:
                if model == 'ghp':
                    while True:
                        y_sim = np.random.poisson(mu_t)
                        if y_sim > 0:
                            break
                elif model == 'ghnb':
                    p_nb = r / (r + mu_t)
                    while True:
                        y_sim = np.random.negative_binomial(r, p_nb)
                        if y_sim > 0:
                            break

            forecasts[i, h] = y_sim

            if y_sim == 0:
                score_pi = 1 - pi_t
                score_mu = -mu_t if model == 'ghp' else 0.0
            else:
                score_pi = -pi_t
                if model == 'ghp':
                    trunc = mu_t * np.exp(-mu_t) / (1 - np.exp(-mu_t))
                    score_mu = y_sim - trunc
                elif model == 'ghnb':
                    den = 1 - (1 + mu_t / r) ** (-r)
                    num = mu_t * (1 + mu_t / r) ** (-(r + 1))
                    score_mu = (y_sim - num) / den

            f1_t1 = omega1 + alpha1 * score_mu + beta1 * f1_t
            f2_t1 = omega2 + alpha2 * score_pi + beta2 * f2_t

            f1_t = np.clip(f1_t1, *logmu_bounds)
            f2_t = np.clip(f2_t1, *logitpi_bounds)

    forecast_mean = np.mean(forecasts, axis=0)
    forecast_ic = np.percentile(forecasts, [2.5, 97.5], axis=0)

    return forecast_mean, forecast_ic


forecast_mean, forecast_ic = forecast_gas_model(
    f1_last=f1_hat[-1],
    f2_last=f2_hat[-1],
    params=parametros_estimados,
    k=10,
    n_sim=5000,
    model='ghp'  # ou 'ghnb'
)
```

```python
def plot_forecast(y_obs, forecast_mean, forecast_ic, title="Previsão k-passos à frente"):
    """
    Plota série observada e previsão k-passos com intervalo de confiança.
    """
    T = len(y_obs)
    k = len(forecast_mean)
    forecast_idx = np.arange(T, T + k)

    plt.figure(figsize=(12, 5))
    plt.plot(np.arange(T), y_obs, label='Série Observada', lw=1.5, color='black')
    plt.plot(forecast_idx, forecast_mean, label='Previsão', lw=2, color='tab:blue')
    plt.fill_between(forecast_idx, forecast_ic[0], forecast_ic[1], color='tab:blue', alpha=0.3, label='IC 95%')

    plt.axvline(T - 1, color='grey', linestyle='--', lw=1)
    plt.xlabel("Tempo")
    plt.ylabel("Contagem")
    plt.title(title)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()


def plot_diagnostics(pit, quant_resid, pearson_resid, title='Diagnóstico do Modelo'):
    """
    Plota gráficos de diagnóstico: PIT, resíduos quantílicos e resíduos de Pearson.
    """
    T = len(pit)

    fig, axs = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle(f'{title} – PIT e Resíduos', fontsize=16)

    sns.histplot(pit, bins=20, kde=False, ax=axs[0, 0], color='skyblue')
    axs[0, 0].set_title('Histograma do PIT')
    axs[0, 0].axhline(T / 20, color='red', linestyle='--')

    sns.histplot(quant_resid, bins=30, kde=True, ax=axs[0, 1])
    axs[0, 1].set_title('Resíduos Quantílicos')

    probplot(quant_resid, dist="norm", plot=axs[1, 0])
    axs[1, 0].set_title('QQ-Plot do Resíduo Quantílico')

    axs[1, 1].scatter(range(T), pearson_resid, alpha=0.7)
    axs[1, 1].set_title('Resíduos de Pearson')

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
```
