- Distribuição condicional
$$
\begin{align}
f(y_t \mid \mathcal{F}_{t-1}) 
&= (1 - \pi_t)\,\mathbb{1}_{\{y_t = 0\}} 
+ \pi_t \, g(y_t \mid \mathcal{F}_{t-1}) \, \mathbb{1}_{\{y_t > 0\}}
\end{align}
$$
$$
\begin{align}
g(y_t \mid \mathcal{F}_{t-1}) \sim \text{Gamma}(\alpha_t, \lambda_t/\alpha_t)
\end{align}
$$

$$
\begin{align}
0 < \pi_t < 1, \quad \lambda_t > 0, \quad \alpha_t > 0
\end{align}
$$

- Média condicional
$$
\begin{align}
\mathbb{E}[y_t \mid y_t > 0] = \lambda_{t+1|t}
\end{align}
$$

$$
\begin{align}
\lambda_{t+1|t} = e^{\tilde{\lambda}_{t+1|t}}
\end{align}
$$

$$
\begin{align}
\tilde{\lambda}_{t+1|t} = \mu_{t+1|t} + \gamma_{t+1|t}
\end{align}
$$

$$
\begin{align}
\mu_{t+1|t}
=
\omega_\mu + \phi_\mu \mu_{t|t-1} + k_\mu \tilde{s}_t
\end{align}
$$

- Sazonalidade trigonométrica
$$
\begin{align}
\gamma_{t+1|t} = \sum_{j=1}^{J} \gamma_{j,t+1|t}
\end{align}
$$

$$
\begin{align}
\begin{pmatrix}
\gamma_{j,t+1|t} \\
\gamma^*_{j,t+1|t}
\end{pmatrix}
=
\begin{pmatrix}
\cos(\lambda_j) & \sin(\lambda_j) \\
-\sin(\lambda_j) & \cos(\lambda_j)
\end{pmatrix}
\begin{pmatrix}
\gamma_{j,t} \\
\gamma^*_{j,t}
\end{pmatrix}
+
\begin{pmatrix}
K_\gamma \\
K_\gamma
\end{pmatrix}
\tilde{s}_t
\end{align}
$$

- Pi
$$
\begin{align}
\pi_{t+1|t} = \frac{1}{1 + e^{-\tilde{\pi}_{t+1|t}}}
\end{align}
$$

$$
\begin{align}
\tilde{\pi}_{t+1|t}
=
a_0 + a_1 \tilde{\lambda}_{t+1|t}
\end{align}
$$

- Alpha
$$
\begin{align}
\alpha_{t+1|t} = e^{\tilde{\alpha}_{t+1|t}}
\end{align}
$$

$$
\begin{align}
\tilde{\alpha}_{t+1|t}
=
\omega_\alpha + \phi_\alpha \tilde{\alpha}_t + K_\alpha \tilde{s}_t
\end{align}
$$
